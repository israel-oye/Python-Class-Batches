```python
#Creating a class and constructor

class Account:
    def __init__(self, name, account_number):
        self.name= name
        self.account_number=account_number
        self.balance= 0
        
    # creating methods
    # creating deposit method
    
    def deposit(self, amount):
        self.balance+=amount
        print(f"{self.name} deposited ${amount}. Current balance is: ${self.balance}")
        
     # creating withdraw method
    
    def withdraw(self, amount):
        if self.balance>=amount:
            self.balance-=amount
            print(f"{self.name} withdraw ${amount}. Current balance is: ${self.balance}")
        else:
            print("You dont have enough money in your account.")
            
            
            
class Savings_Account(Account):
    def __init__(self, name, account_number,interest_rate):
        super().__init__(name,account_number)
        self.interest_rate=interest_rate
        
    # add interest method
    
    def add_interest(self):
        interest = self.balance*self.interest_rate
        self.deposit(interest)
        
        
        
#Creating and Using the objects

acc1 = Account("peter", "123456")
acc1.deposit(500)
acc1.withdraw(450)

print()

savings_account=Savings_Account("patrick","124564", 0.05)
savings_account.deposit(1000)
savings_account.add_interest()
savings_account.withdraw(500)
savings_account.withdraw(1000)
        

    
```

```python
# Operator overloading 

class Hope:
    def __init__(self,x,y):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f"({self.x},{self.y})"   
    
    def __add__(self, anotherObject):
        r1 = self.x + anotherObject.x
        r2 = self.y + anotherObject.y
        return (r1,r2)
        
obj1 = Hope(4,10)   
obj2 = Hope(5,6)   

print(obj1 + obj2) # Hope.__add__(obj1, obj2)
#print(obj1*obj2)
#print(obj2)

#print(Hope.__add__(obj1,obj2))
```

```python
# Operator overloading 

class Combine:
    def __init__(self,x,y):
        self.x = x
        self.y = y   
    
    def __gt__(self,other): 
        r1 = self.x + self.y
        r2 = other.x + other.y
        
        if r1>r2:
            return True
        else:
            return False
        
           
obj1 = Combine(2,10)  #32
obj2 = Combine(5,15)  #20

#print(obj1 + obj2)
#Combine.__add__(obj1,obj2)

if obj1>obj2:
    print("waooooooo")
else:
    print("beautiful")
```

```python
"""
abstraction(abstract class and abstract method)
Note: we can't create the object of an absract class, to create the object, we need to create 
a class that will inherit it. For a class to be an abstract class, it must have at least
one abstract method. Abstract method will not have body or definition inside its class, instead, we need to define it inside
the class that inherit the abstract class.
 
"""

from abc import ABC, abstractmethod            # ABC(Abstract Base Class)

class Computer(ABC):
    
    @abstractmethod
    def show(self):
        pass

    def show1(self):
        print("not an abstract method")
    
    
class Laptop(Computer):
    def show(self):
        print("hellooo abstract")


#c = Computer()

lap1 = Laptop()
lap1.show1()
```

```python
# Note: we can have more than one abstract method inside an abstract class.
from abc import ABC, abstractmethod            # ABC(Abstract Base Class)

class Computer1(ABC):
    
    @abstractmethod
    def show(self):
        pass

    @abstractmethod    
    def show1(self):
        pass
    
    def show2(self):
        print("not an abstract method")
    
    
class Laptop1(Computer1):
    def show(self):
        print("hellooo abstract method1")
    def show1(self):
        print("hellooo abstract method1")


#c = Computer()

lap1 = Laptop1()
lap1.show()
```

```python
class Book:
    
    def __init__(self):
        self.name ="Harrypotter" # public variable(we can access it anywhere)
        
        self._page= 24  # protected variable(can only be accessed inside its class and the class that inherit its class)
        self.__author = "jack"   # private variable(can only be accessed inside its class)
    
    def display(self):
        print("this is the first display method",self.__author)
    
    def _display1(self):
        print("this is the second display method",self.__author)
    
    def __display2(self):
        print("this is a third display method",self.__author)
        
b = Book()
print(b.name)


```

```python
# Multithreading

from threading import Thread
from time import sleep

class Test1(Thread):
    def run(self):
        for i in range(5):
            print("helooo")
            sleep(1)

class Test2(Thread):
    def run(self):
        for i in range(5):
            print("hiii")
            sleep(1)
t1 = Test1()
t2 = Test2()

t1.start()
sleep(0.5)
t2.start()

t1.join()
t2.join()


print("GoodBye")

# Note: join method will allow the main thread to wait untill the two threads 
# complete their executions.
```

```python
class Employee:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city
        
    def show(self):
        print(f"Employee's Name: {self.name}, Employee's age:{self.age}, Employee's city:{self.city}")
        
emp1 = Employee("John", 25, "Lagos")

emp1.show()
                       
```

```python
# Unserialization or Unpickling

with open('info.pickle', 'rb') as f:
    emp = pickle.load(f)
    
emp.show()
```

```python
import threading
import time

def print_letters(letters: str):
    for char in letters:
        print(char)
    time.sleep(2)


def print_nums(upper_bound: int):
    for n in range(1, upper_bound):
        print(n)


thread1 = threading.Thread(target=print_letters, args=('qwerty',))
thread2 = threading.Thread(target=print_nums, args=(10,))

thread1.start()
thread2.start()

thread1.join()
thread2.join()
```

```python
import threading


class NameThread(threading.Thread):
    names = ['Zack', 'Yucatan', 'Xi', 'Wuraola', 'Veronica']
    def run(self):
        for name in NameThread.names:
            time.sleep(2)
            print(name)

class NumberThread(threading.Thread):
    def run(self):
        for i in range(5):
            time.sleep(2)
            print(i)


thread1 = NameThread()
thread2 = NumberThread()

thread1.start()
thread2.start()

thread1.join()
thread2.join()
```
