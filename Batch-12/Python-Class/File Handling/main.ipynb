```python
with open('fruits.txt', 'r') as f:
    lines = f.readlines()
    for line in lines:
        print(line.strip().upper())
```

```python
with open('poem.txt', 'r') as f:
    lines = f.readlines()
    f.readline()
    wordful_lines = [line for line in lines if line.strip()]
    print(f"Poem in the file has {len(wordful_lines)} lines")
```
