```python
import tkinter as tk # alias importing


window = tk.Tk()
# window.geometry('300x200')
window.title("My App")
window.minsize(width=400, height=300)

frame = tk.Frame(window)

# Label
label = tk.Label(frame, text='Welcome', font=('Arial', 24, 'italic'))
label.pack()

# Entry
textbox = tk.Entry(frame)
textbox.pack()

result_label = tk.Label()
result_label.pack()

def on_click():
    user_typed = textbox.get()
    result_label.config(text=user_typed)
    textbox.delete(0, tk.END)

# Button
btn = tk.Button(frame, text='Click', command=on_click)
btn.pack()

frame.pack()
window.mainloop()
```

```python
import tkinter as tk


root = tk.Tk()
root.title("Temperature Converter")
root.minsize(width="300", height="200")


app_label = tk.Label(text="°C to °F")
input_box = tk.Entry()
result_label = tk.Label()


def on_submit():
    temp_in_c = float(input_box.get())
    input_box.delete(0, tk.END)
    
    temp_in_f = temp_in_c * 1.8 + 32

    result_label.config(text=f"{temp_in_f:.2f}°F")

submit_btn = tk.Button(text="Convert", command=on_submit)

app_label.pack()
input_box.pack()
submit_btn.pack()
result_label.pack()


root.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.title("Login Form")
root.minsize(width=300, height=250)

username = 'john.doe'
password = 'JohnD03'

# Appl label
app_lbl = tk.Label(root, text='Login', font=('Sans-Serif', 12, 'bold'))
app_lbl.pack(side='top')


msg_lbl = tk.Label(root, text='')
msg_lbl.config(pady=24)
# msg_lbl.pack(side='top')

# Form frame
frame = tk.Frame(root, width=250, height=200)
frame.config(bg='snow2', padx=12, pady=16)
frame.pack(expand=True)
frame.grid_columnconfigure(0, weight=0, pad=4)
frame.grid_columnconfigure(1, weight=1)


# Form labels
username_lbl = tk.Label(frame, text='Username', font=('Arial', 10,))
password_lbl = tk.Label(frame, text='Password', font=('Arial', 10))

# Form entries
username_entry = tk.Entry(frame)
password_entry = tk.Entry(frame, show='•')


def on_click():
    if messagebox.askyesno("Confirm", "Are you sure you want to login?"):
        on_submit()
        return
    # messagebox.
    return

def on_submit():
    if username_entry.get() == username and password_entry.get() == password:
        # msg_lbl.config(text="Login sucess ✅")
        messagebox.showinfo('Login success', message="Successfully logged in", parent=root)
        username_entry.delete(0, tk.END)
        password_entry.delete(0, tk.END)
    else:
        messagebox.showerror('Invalid details', "Invalid username/password given", parent=root)
        # msg_lbl.config(text="Invalid username/password ❌")
    

# Form button
submit_btn = tk.Button(frame, text='Submit', command=on_click)
submit_btn.config(bg="steel blue", fg="white", padx=4, pady=6, relief='raised')


# Layout of form
username_lbl.grid(column=0, row=0, padx=8)
password_lbl.grid(column=0, row=1)
username_entry.grid(column=1, row=0)
password_entry.grid(column=1, row=1)
submit_btn.grid(column=0, row=2, columnspan=2, pady=12, sticky='s')


msg_lbl.pack()
root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.title("Menu App")
root.geometry("300x250")

menubar = tk.Menu(root)

def open_window():
    new_window = tk.Toplevel(root)
    new_window.title("New Window")
    


file_menu = tk.Menu(menubar, tearoff=False)
file_menu.add_command(label="New text file", command=lambda: print("created new text file"))
file_menu.add_command(label="New window", command=open_window)
file_menu.add_command(label="New file")

open_submenu = tk.Menu(file_menu)
open_submenu.add_command(label="Window")
open_submenu.add_command(label="File")


file_menu.add_cascade(label='Open', menu=open_submenu)
file_menu.add_command(label='Quit', )

edit_menu = tk.Menu(menubar)
selection_menu = tk.Menu(menubar)
view_menu = tk.Menu(menubar, tearoff=False)

dark_mode = tk.BooleanVar(value=False)

def toggle():
    print(dark_mode.get())
    if dark_mode.get():
        root.config(bg='grey52')
        return
    root.config(bg='grey99')

view_menu.add_checkbutton(label='Dark Mode', variable=dark_mode, command=toggle)


menubar.add_cascade(label="File", menu=file_menu)
menubar.add_cascade(label="Edit", menu=edit_menu)
menubar.add_cascade(label="Selection", menu=selection_menu)
menubar.add_cascade(label='View', menu=view_menu)

root.config(menu=menubar)
root.mainloop()
```

```python
import tkinter as tk

class MyApp(tk.Tk):
    def __init__(self, screenName = None, baseName = None, className = "Tk", useTk = True, sync = False, use = None):
        super().__init__(screenName, baseName, className, useTk, sync, use)
        self.title("My App")
        self.geometry("300x250")
        self.__count = 0
        self.setup_widgets()

    def __increment(self):
        self.__count += 1
        self.label.config(text=f'{self.__count}')

    def __decrement(self):
        self.__count -= 1
        self.label.config(text=str(self.__count))

    def run(self):
        self.mainloop()
    
    def setup_widgets(self):
        self.frame = tk.Frame(self)
        self.frame.grid_columnconfigure(3, weight=3)
        self.frame.pack(expand=True)


        self.label = tk.Label(self.frame, text=f'{self.__count}', font=('Arial', 24, 'bold'))
        
        self.add_btn = tk.Button(self.frame, text='+', command=self.__increment)
        self.add_btn.config(bg='SeaGreen1', fg='grey5', padx=6, pady=6, font=('sans-serif', 12, 'bold'))

        self.minus_btn = tk.Button(self.frame, text='-', command=self.__decrement)
        self.minus_btn.config(bg='Tomato3', fg='grey99', padx=6, pady=6, font=('sans-serif', 12, 'bold'))

        self.label.grid(row=1, column=3)
        self.add_btn.grid(row=3, column=1)
        self.minus_btn.grid(row=3, column=5)



app = MyApp()
app.run()
```

```python
import re

def count_vowels(text: str):
    vowels = re.findall(r"[aeiou]", text)
    return len(vowels)

count_vowels(text="Lorem ipsum")
```

**Exercise 4 — Sanitise a Form Field**

A username field should only contain letters, digits, and underscores. Use `re.sub()` to strip any invalid characters from `"lumi@scala#ware$2024!"`. Expected: `"lumiscalaWare2024"` (or similar).

```python
username = "lumi@scala#ware$2024!"

re.sub(r"[@#$!]", "", username)

def sanitize_username(username: str):
    return re.sub(r'[@#$%^&!]', '', username)
```
