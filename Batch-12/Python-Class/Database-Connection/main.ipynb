```python
import sqlite3

connection = sqlite3.connect("data.db")

cursor = connection.cursor()

def create_tables():
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY,
        name TEXT,
        email TEXT
        );       
        """
    )

def populate_db(name: str, email: str):
    try:
        cursor.execute(f"""
        INSERT INTO users (name, email) VALUES (?, ?);
        """,
        (name, email,)
        )
    except sqlite3.Error as e:
        print("An error occured while executing the query ❌", e)
    else:
        connection.commit()
        print("Populate success... ✅")

def delete_user(email: str):
    query = "DELETE FROM users WHERE email = ?"
    cursor.execute(query, (email,))
    connection.commit()


create_tables()
new_username = input("Enter user name").strip().title()
new_user_email = input("Enter email").strip().lower()

if new_user_email and new_username:
    populate_db(name=new_username, email=new_user_email)
else:
    print("Invalid email/username")
delete_user(email='ajoke.babs@gmail.com')
```

```python
import sqlite3

connection = sqlite3.connect('data.db')
cursor = connection.cursor()

def create_table():
    global cursor

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS users(
            id INTEGER PRIMARY KEY,
            name TEXT,
            username TEXT,
            email TEXT UNIQUE,
            address TEXT,
            company TEXT         
        )
    """)


def deserialize_json():
    import json

    with open('users.json', 'r') as f:
        all_users = json.load(f)
    return all_users

def save_user(user: dict):
    global cursor

    _user_address = user.get('address')
    full_address = (
        f"{_user_address['suite']}, {_user_address['street']}, {_user_address['city']}"
    )

    cursor.execute("""
        INSERT INTO users (name, username, email, address, company)
        VALUES (?, ?, ?, ?, ?)
        """,
        (user['name'], user['username'], user['email'], full_address, user['company']['name'])
    )

    connection.commit()

def populate_users_table(users: list[dict]):
    for user in users:
        save_user(user=user)
    print("Finished ✅")



class User:
    def __init__(self, id, name, username, email, address, company):
        self.id = id
        self.name = name
        self.username = username
        self.email = email
        self.address = address
        self.company = company

    @classmethod
    def from_tuple(cls, data: tuple):
        return cls(*data)





def fetch_user(id: int):
    global cursor

    cursor.execute("""
        SELECT * FROM users WHERE id = ?
        """,
        (id,)
    )
    record = cursor.fetchone()
    if record is None:
        print(f"No user with id: {id}")
    else: 
        print(record)
        user = User.from_tuple(record)
        print(user.name, user.id)

    cursor.close()


def main():
    create_table()
    # all_users = deserialize_json()
    # populate_users_table(users=all_users)
    fetch_user(id=10)


if __name__ == '__main__':
    main()
```
