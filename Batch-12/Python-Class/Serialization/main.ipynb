```python
import json

with open('app.json', 'r') as f:
    app_dict = json.load(f)
    
def get_transaction(txn_id: str):
    for txn in app_dict['recent_transactions']:
        if txn['id'] == txn_id:
            return txn
        return None
    

lookup_id = input("Enter transaction id: ")
print(get_transaction(txn_id=lookup_id))
```
