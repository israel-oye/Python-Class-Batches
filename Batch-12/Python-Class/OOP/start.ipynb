```python
class Student:
    """Student class"""

    def __init__(self, id: int, name: str, scores: list[float] = []):
        self.id = id
        self.name = name
        self.scores = scores

    def get_max_score(self):
        if self.scores:
            return max(self.scores)
        return None

    def add_score(self, score: float | int):
        self.scores.append(score)

student1 = Student(1, "Ayo")
student2 = Student(2, "Bola", [9, 10, 8.5])
student3 = Student(id=3, name="Curtis", scores=[9, 9, 3])


# for stud in [student1, student2, student3]:
#     print(stud.name, stud.scores)

#     print(stud.get_max_score())

student4 = Student(4, "Deborah", [5, 7.5, 6.5])
print(student4.get_max_score())

student4.add_score(8)
print(student4.scores) # output -> [5, 7.5, 6.5, 8]
```

```python
class Dog:
    def __init__(self, name: str, breed: str):
        self.name = name
        self.breed = breed

    def bark(self):
        print(f"{self.name} says woof!")

    def rename(self, new_name: str):
        self.name = new_name

    def describe(self):
        print(f"I'm a {self.breed} dog named {self.name}")


bingo = Dog('Bingo', 'Basenji')
bingo.bark()
bingo.rename("Jack")
bingo.bark()
bingo.describe()

bullet = Dog(name='Bullet', breed='Rotweiler')
bullet.rename('Max')
bullet.describe()
```
