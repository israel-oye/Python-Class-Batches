```python
class Student:
    """Student class"""

    def __init__(self, id: int, name: str, scores: list[float] = []):
        self.id = id
        self.name = name
        self.scores = scores

    def get_max_score(self):
        if self.scores:
            return max(self.scores)
        return None

    def add_score(self, score: float | int):
        self.scores.append(score)

student1 = Student(1, "Ayo")
student2 = Student(2, "Bola", [9, 10, 8.5])
student3 = Student(id=3, name="Curtis", scores=[9, 9, 3])


# for stud in [student1, student2, student3]:
#     print(stud.name, stud.scores)

#     print(stud.get_max_score())

student4 = Student(4, "Deborah", [5, 7.5, 6.5])
print(student4.get_max_score())

student4.add_score(8)
print(student4.scores) # output -> [5, 7.5, 6.5, 8]
```

```python
class Dog:
    def __init__(self, name: str, breed: str):
        self.name = name
        self.breed = breed

    def bark(self):
        print(f"{self.name} says woof!")

    def rename(self, new_name: str):
        self.name = new_name

    def describe(self):
        print(f"I'm a {self.breed} dog named {self.name}")


bingo = Dog("Bingo", "Basenji")
bingo.bark()
bingo.rename("Jack")
bingo.bark()
bingo.describe()

bullet = Dog(name="Bullet", breed="Rotweiler")
bullet.rename("Max")
bullet.describe()
```

```python
class Circle:
    def __init__(self, radius: float):
        self.radius = radius

    def area(self):
        import math
        return math.pi * (self.radius ** 2)

    def get_area_dp(self, places: int):
        return round(self.area(), places)


circle1 = Circle(radius=14)
print(circle1.area(), circle1.get_area_dp(0))

circle2 = Circle(20)
print(circle2.area())
```

```python
class Triangle:
    def __init__(self, base: float, height: float):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.height * self.base
    

class Rectangle:
    def __init__(self, length: float, width: float):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
    
    def perimeter(self):
        return 2 * (self.length + self.width)
    
    def get_area_2dp(self):
        return f"{self.area():.2f}"

rect1 = Rectangle(5, 4)
print(rect1.area(), rect1.perimeter())

rect2 = Rectangle(10, 8)
print(rect2.area(), rect2.perimeter())
```

```python
class Circle:
    def __init__(self, d: float):
        self.diameter = d

    def area(self):
        import math
        return round((math.pi * (self.diameter / 2) ** 2), 2)

    @classmethod
    def from_radius(cls, radius: float):
        _diameter = radius * 2
        new_circle = cls(d=_diameter) #Circle(d=_diamater)
        return new_circle

circle1 = Circle(d=17) # Circle.__init__(d=17)
print(circle1.area())

circle2 = Circle.from_radius(8.5)
print(circle2.area())
```

```python
class User:
    """User class"""

    def __init__(self, id: int, name: str):
        self.id = id
        self.name = name

    def describe(self):
        print(f"Hello. My name is {self.name.title()}")

    @classmethod
    def anonymous(cls):
        return cls(0, 'Anonymous')
    
    

user1 = User(1, 'Abeebi')
anon = User.anonymous()

user1.describe()
anon.describe()
print(anon.id)
```

```python
class User:
    def __init__(self, username: str, email: str):
        self.username = username
        self.email = email

    @classmethod
    def from_string(cls, user_detail: str):
        details = user_detail.split(',')
        username = details[0]
        email = details[1].lower()

        return cls(username, email)


user_1 = User("hardey01", 'ademola@gmail.com')
user_2 = User.from_string("smiling-j0e,johannu.oba@yahoo.com")

print(user_1.username, user_2.username)

import time
for user in [user_1, user_2]:
    print(f"Sending email to {user.email}")
    time.sleep(1.5)
```
