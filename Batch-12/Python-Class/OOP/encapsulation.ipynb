```python
class Locker:
    def __init__(self):
        self.__pin = '1234'

    def check_pin(self, pin: str):
        if self.__pin == pin:
            print("PIN is correct")
        else:
            print("Incorrect PIN")

    @property
    def pin(self):
        return self.__pin
    
    @pin.setter
    def pin(self, value):
        self.__pin = value

    def set_pin(self, value: str):
        if value.isdigit() and len(value) == 4:
            self.__pin = value
        else:
            print("Pin must be 4 digits")


my_locker = Locker()

# my_locker.check_pin("0012")
# my_locker.set_pin("ABCD")
# my_locker.check_pin("ABCD")
# my_locker.check_pin("1234")

my_locker.pin = '9900'
my_locker.check_pin('9900')
```

```python
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def __str__(self):
        return f"(Rectangle l: {self.length} | b: {self.breadth})"


class Square(Rectangle):
    def __init__(self, side):
        super().__init__(length=side, breadth=side)


square = Square(5)
print(square)
print(square.breadth, square.length)
```

```python
class LivingThing:
    def respiration(self):
        print("Breathing...")

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def describe(self):
        print(f"My name is {self.name}")


class Student(Person, LivingThing):
    def __init__(self, name, age):
        super().__init__(name, age)

    def describe(self):
        # Method overriding
        print(f"Student\nName: {self.name}")

student = Student('Emma', 45)
print(Student.mro())
student.describe()
student.respiration()

# MRO: Method Resolution Order
```

```python
class A:
    def func(self):
        return "A"
    
class B:
    def func(self):
        return "B"
    
class C(A):
    def func(self):
        return "C " + super().func()
    
    def __str__(self):
        return f"(C object at {id(self)})"
    
c = C()
c.func()
print(c)
```

```python
#  Abstraction
from abc import abstractmethod, ABC

class Animal(ABC):
    @abstractmethod
    def eat(self):
        pass

    @abstractmethod
    def move(self):
        pass


class Goat(Animal):
    def eat(self):
        return "Eating plant-based food"

    def move(self):
        return "Walks on land..."

class Dog(Animal):
    pass

bingo = Dog()


class Shark(Animal):
    def eat(self):
        return "Eating fish and flesh..."

    def move(self):
        return "Swimming in the ocean..."


def get_animal_behaviour(animal: Animal):
    print(f"This animal eats like: {animal.eat()}\nAnd moves like: {animal.move()}")


goat = Goat()
shark = Shark()

print(type(goat))
print(isinstance("abc", int))
print(isinstance(goat, Animal))

# get_animal_behaviour(goat)
```

```python
from abc import abstractmethod, ABC

class Weapon(ABC):
    @abstractmethod
    def use(self):
        pass

class Sword(Weapon):
    def use(self):
        print("Fighting with sword...")


class Knife(Weapon):
    def use(self):
        print("Fighting with knife...")


class Gun(Weapon):
    def use(self):
        print("Fighting with gun...")


class Character(ABC):
    @abstractmethod
    def describe(self):
        pass

    def fight(self, weapon: Weapon):
        weapon.use()


class King(Character):
    def describe(self):
        print("I am a king character...")

class Queen(Character):
    def describe(self):
        print("I am a queen character...")

class Soldier(Character):
    def describe(self):
        print("I am a soldier character...")


sword = Sword()
knife = Knife()
gun = Gun()

king = King()
king.fight(sword)
```
