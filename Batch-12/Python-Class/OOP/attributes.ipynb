````python
from decimal import Decimal

class BankAccount:
    bank_name = "National Trust Bank"

    def __init__(self, account_holder: str, account_num: str):
        self.account_holder = account_holder
        self.account_number = account_num
        self.balance = Decimal('0.00')

    def deposit(self, amount: str):
        amount = Decimal(amount)
        self.balance += amount

    def withdraw(self, amount: str):
        amount = Decimal(amount)
        if amount > self.balance:
            print("Insufficient funds...")
            return
        self.balance -= amount
        print(f"Withdrawn: {amount}\t Bal: ₦{self.balance}")

    def get_balance(self):
        return f"₦{self.balance}"

    def bal_in_usd(self):
        return f"${self.balance / Decimal(1384.30):.2f}"


account1 = BankAccount("Alice Johnson", "ACC001")
account2 = BankAccount("Bob Smith", "ACC002")

# Access class attribute both ways
print(BankAccount.bank_name)
print(account1.bank_name)

# Display information and perform operations
# account1.display_account_info()
account1.deposit(1500)
account1.withdraw(100)

# Compare balances between accounts
if account1.balance > account2.balance:
    print(f"{account1.account_holder} has more money")

print(account1.get_balance(), account2.get_balance())
print(account1.bal_in_usd())

# Add an instance method called get_balance_in_usd() that converts the balance
# to the USD equivalent
# $1 == ₦1,384.30k
# Add an instance method get_balance_in_currency that converts the account's balance to
# the entered currency
# account.get_balance_in_currency('Euros') -> €1.50
# account.get_balance_in_currency('pounds') -> £1.45
# account.get_balance_in_currency('yen') -> ¥0.24
# account.get_balance_in_currency('rand') -> N/A
````

````python
class Book:
    library_name = "Osun Library"

    def __init__(self, title: str, author: str, isbn: str, copies: int, pub_year: int):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.copies_available = copies
        self.year_published = pub_year

    def borrow_book_copies(self, copies: int):
        if copies <= self.copies_available:
            self.copies_available -= copies
            print(f"Borrowed {copies} copies\tLeft: {self.copies_available} left")
        else:
            print(f"Invalid request. Cannot borrow more than available")

    def return_book(self):
        self.copies_available += 1

    def borrow_book(self):
        self.copies_available -= 1

    def is_available(self):
        return self.copies_available >= 1

    def book_info(self):
        data = {
            'title': self.title.title(),
            'author': self.author.title(),
            'isbn': self.isbn.upper(),
            'copies': self.copies_available,
            'year': self.year_published,
            'is_available': self.is_available()
        }
        return data

book1 = Book("Python Programming", "John Smith", "ISBN001", 5, 2020)
book2 = Book("Data Science Basics", "Jane Doe", "ISBN002", 2, 2021)
book3 = Book("Web Development Guide", "Mike Johnson", "ISBN003", 8, 2019)

# Access class attribute both ways
print(Book.library_name)
print(book1.library_name)

# Display information and perform operations
print(book1.book_info())
book1.borrow_book()
book1.borrow_book()
print(f"Book 1 available: {book1.is_available()}")
````
