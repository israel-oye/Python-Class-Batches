````python
word = input("Enter word: ").strip().lower()
consonant_count, vowel_count, digit_count = (0, 0, 0)

# DRY - Don't Repeat Yurself
for char in word:
    if char.isalpha() and char in 'aeiou':
        vowel_count += 1
    if char.isalpha() and (char not in 'aeiou'):
        consonant_count += 1
    if char.isdigit():
        digit_count += 1
    

print(f"{word}\nConsonants: {consonant_count}\nVowels: {vowel_count}\nDigits: {digit_count}")
````

*Holiday Project 1:* Password Generator Program

*Objective:*
The goal of this exercise is to help you practice working with lists, user input, loops, and basic string manipulation. You will build a simple password generator that creates a random password based on user-defined criteria.

*Description:*
You are required to develop a program that generates a password for a user. The program should first ask the user to specify:

- the number of alphabetic characters they want in the password  
- the number of digits they want  
- the number of special symbols they want  

Based on the user’s input, your program should generate a password that contains exactly those quantities.

*Requirements:*
- Create *three separate lists of strings*:  
  - one list containing alphabets (uppercase and/or lowercase letters),  
  - one list containing digits (`0–9`),  
  - one list containing special symbols (for example: `! @ # $ % & *`).  
- Prompt the user to enter how many characters they want from each category.  
- Randomly select characters from each list according to the user’s input.  
- Combine the selected characters to form a single password.  
- Display the generated password to the user.  

*Rules and Constraints:*
- The password must match the exact counts entered by the user.  
- Characters should be chosen randomly, not in a fixed or predictable order.  
- Do not hard-code the final password.  
- Use clear variable names and readable code.  

*Expected Output:*  
After the user provides their inputs, the program should print a randomly generated password that satisfies the given conditions.

````python
import random #absolute import
from string import ascii_letters, digits #selective imports

alphabet_list = list(ascii_letters)
digit_list = list(digits)
symbols_list = list("!@#$%&*")


char_len = int(input("How many alphabets do you want? ")) #4
digits_len = int(input("How many digits do you want? "))
symbols_len = int(input("How many special chars do you want? "))

password = ""

for _ in range(char_len):
    password += random.choice(alphabet_list)

for _ in range(digits_len):
    password += random.choice(digit_list)

for _ in range(symbols_len):
    password += random.choice(symbols_list)


print(f"Generated password: {password}")

password_list = list(password)
random.shuffle(password_list)

final_password = "".join(password_list)

print(f"Final password: {final_password}")
````

````python
# Simulate an ATM machine with a while-loop
# You should check if the user pin is correct before starting the machine
# Allow the user to enter a menu option to perform operations on the machine.
# e.g user-inputs 1. This should take them to the check balance screen.
# Write the appropriate code for each menu

balance = 100
pin = "1234"

def get_balance():
    return balance

def deposit(amount: float):
    balance += amount
    print(f"Deposited ${amount}. Bal: ${balance}")

def withdraw(withdraw_amount: float):
    if withdraw_amount > balance:
        print("Insufficient funds.\n")
    else:
        balance -= withdraw_amount
        print(f"Debit: {withdraw_amount}. Bal: {balance}\n")


def display_atm():
    while True:
        print("\n1. Check Balance\n2. Deposit\n3. Withdraw\n4. Exit")
        choice = input("Choose an option: ")

        if choice == "1":
            print(f"Bal: ${get_balance()}\n")
        elif choice == "2":
            deposit_amount = float(input("How much do you want to deposit: "))
            deposit(amount=deposit_amount)
        elif choice == "3":
            withdraw_amount = float(input("How much to withdraw?: "))
            withdraw(withdraw_amount)
        elif choice == "4":
            confirm = input("Are you sure you want to exit ATM (Y/N): ").lower()
            if confirm == 'y':
                print("Thanks for banking with us...\n")
                break
        else:
            print("Invalid choice")

entered_pin = input("Enter PIN: ")

if entered_pin != pin:
    print("Incorrect PIN! Exiting...")
else:
    display_atm()
````

````python
def math_operation(operation: str, *operands):
    operation = operation.lower()
    if operation == 'add':
        return sum(operands)
    elif operation == 'multiply':
        product = 1
        for n in operands:
            product = product * n
        return product
    elif operation == 'average':
        return sum(operands) / len(operands)
    elif operation == 'subtraction' or operation == 'sub':
        diff = operands[0]
        for n in operands[1:]:
            diff -= n
        return diff
    else:
        print("Invalid operation")
    

# math_operation('ADD', 70, 50)
# math_operation('average', 3, 5, 6, 10)
# math_operation('multiply', 2, 3, 1, 4)
math_operation('sub', 6, 6, 1)
````

## Decorators in Python: An Intuitive Explanation with a Restaurant Analogy

### The Restaurant Analogy

Imagine you’re at a restaurant, and you order a dish. The dish represents a **function** in Python — it’s the core thing you want. However, sometimes you want that dish to have extra things on it, like a topping, garnish, or sauce. These extra things are **decorations** — they enhance the dish without changing the core meal. Similarly, a **decorator** in Python is like a way of adding extra functionality to a function (your dish) without changing its core logic.

Here’s how we can break this analogy into concrete parts:

---

1. **The Basic Dish (Function)**

   * You have a dish (function) that’s good on its own, say a **burger** (the basic function).
   * If you just make the burger without anything extra, you’re using the function as-is.

2. **The Chef’s Topping (Decorator)**

   * The restaurant's chef can add special toppings to the burger, like lettuce, tomato, and a fancy sauce, making the burger even better. This is the decorator.
   * The **decorator** doesn't change the burger’s essence — it's still a burger — but it adds something to it (like logging, timing, or extra validation).

3. **The Burger with Toppings (Decorated Function)**

   * When you order the burger with those toppings, you get the original burger but with enhancements.
   * In Python, when you apply a decorator to a function, you get a new version of that function with the added features from the decorator.

---

### Code Example:

Let’s say we have a simple function that prints a greeting. You can think of it like the **plain burger**.

```python
def greet():
    print("Hello!")
```

Now, imagine we want to add a special “greeting time” feature, so we know when the greeting happens (just like adding toppings to the burger).

#### Without a Decorator (Manual Change):

You could manually change the function by adding time tracking inside it, like this:

```python
import time

def greet():
    start_time = time.time()
    print("Hello!")
    print(f"Greeted in {time.time() - start_time} seconds.")
```

But this makes the original function more complex and changes it directly.

#### With a Decorator (Keeping it Simple):

A decorator is a cleaner way to add this extra behavior without changing the original function. Here's how the decorator would work:

```python
import time

def timing_decorator(func):
    def wrapper():
        start_time = time.time()
        func()
        print(f"Greeted in {time.time() - start_time} seconds.")
    return wrapper

@timing_decorator
def greet():
    print("Hello!")
```

Now the `greet()` function is still the same (a plain burger), but we've added a **timing feature** (the special toppings) using the `@timing_decorator`. When you call `greet()`, it’s as if you're getting the burger with the toppings:

```python
greet()
# Output:
# Hello!
# Greeted in 0.00234 seconds.
```

---

### Breakdown:

* `greet()` is the original function, the **core dish**.
* `timing_decorator` is the decorator, like the chef’s topping or extra sauce.
* The `@timing_decorator` is like saying “please add the topping to my dish without me having to change it directly.”

---

### Why Decorators are Useful:

1. **Separation of Concerns:** You can add extra functionality like logging, timing, or validation to a function without cluttering its main logic.

2. **Reusability:** You can reuse decorators on multiple functions, just like how you can add similar toppings to many dishes without rewriting them every time.

### Recap:

* A decorator is like an extra feature or topping for a dish (function).
* It doesn't change the original function but enhances it in some way.
* The `@decorator_name` syntax is a shorthand to apply a decorator, making code cleaner and easier to read.

This analogy of “topping your dish” is intuitive because we already know that we can add extras to our food without changing the food itself. Similarly, decorators let us add extra functionality to functions without altering their core behavior.

````python
def upper_decorator(func):
    def wrapper():
        value: str = func()
        return value.upper()
    
    return wrapper


def get_name():
    return "Solomon Grundy"

@upper_decorator
def get_country():
    return "Nigeria"

print(get_country())

get_name_ = upper_decorator(func=get_name)
print(get_name_(), get_name())
````

````python
def make_bold(func):
    def wrapper(arg):
        # f"<b>{get_text(arg)}</b>"
        return f"<b>{func(arg)}</b>"

    return wrapper


def make_italics(func):
    def wrapper():
        return f"<i>{func()}</i>"

    return wrapper


def get_text(text: str):
    return text


decorated_get_text = make_bold(func=get_text)
print(decorated_get_text("xyz"))

@make_bold
def _get_text(text: str):
    return text

_get_text("ABC")
````

````python
balance = 2500.54

user_password = "1234"

def login_required(func):
    def wrapper(*args, **kwargs):
        password = input("Enter password")
        if password == user_password:
            return func(*args, **kwargs)
        print("Unauthorized....")
    
    return wrapper

@login_required
def get_balance():
    global balance
    return balance

@login_required
def get_statement():
    return "Statement of Account.pdf"

@login_required
def change_pin(new_pin: str):
    global user_password
    user_password = new_pin
    print("Password changed...")

@login_required
def add_details(**details):
    print(details)


# print(get_statement())
# print(get_balance())
change_pin(new_pin="0987")
print(get_balance())
add_details(name="Emma", surname="Hammed", gender="Male", is_sleeping=True)
````

````python
def timer_decorator(func):
    def wrapper(*args, **kwargs):
        import time
        start = time.perf_counter()
        func(*args, **kwargs)
        end = time.perf_counter()
        diff = end - start
        print(f"{func.__name__} took {diff:.2f}s")
    return wrapper


@timer_decorator
def count_occurence(sentence: str, word: str):
    word_count = {}

    for _word in sentence.split():
        if _word in word_count:
            word_count[_word] += 1
        else:
            word_count[_word] = 1
    print(word_count)

    frequency = word_count.get(word, 0)
    if not frequency:
        print(f"'{word}' is not found")
        return frequency
    return frequency


count_occurence("Chukwuka is a boy. He is a good boy", "boy")

# @timer_decorator
def count_to_n(n):
    for i in range(1, n + 1):
        pass

count_to_n = timer_decorator(func=count_to_n)
count_to_n(2000)
````

````python
def retry(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@retry(4)
def greet():
    print("Hi")

greet()

# decorator = retry(3)
# wrapper = decorator(func=greet)
# wrapper()
````

````python
def wrap_with_tag(tag: str):
    def decorator(func):
        def wrapper(*args, **kwargs):
            return f"<{tag}>{func(*args, **kwargs)}</{tag}>"
        return wrapper
    return decorator

@wrap_with_tag("h2")
def hello_world():
    return "Hello world"

@wrap_with_tag('p')
def generate_placeholder(word_count: int):
    return "lorem" * word_count

hello_world()
generate_placeholder(5)
````

````python
# Recursion
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

factorial(4)
````

````python
def count_up_gen(n):
    i = 0
    while i <= n:
        yield i
        i += 1

for j in count_up_gen(10_000_000_000):
    print(j)
````

````python
# try:
#    somethng that may cause an error
# except TheErrorThatHappens:
#   do something when the error comes up
# else:
#    do this if no error

def divide(x, y):
    return x / y


def divide_(x, y):
    try:
        result = x / y
    except ZeroDivisionError as exc: #Alias
        print("Sorry you can't divide by 0", exc.__class__.__name__)
    except TypeError:
        print("Can only divide numbers by numbers (int/float)")
    else:
        return result
    

print(divide_(2, 4))
print(divide_(9, 0))
print(divide_('10', 5))
````

````python
def check_email(email: str):
    if '@' in email:
        return True
    else:
        raise ValueError("Invalid email")
    
# check_email('jagun@gmail.com')
# check_email('hammymail.com')

def validator(email):
    try:
        check_email(email)
    except ValueError as exc:
        print(exc)
    else:
        print(f"{email} is valid ✅")


validator('fareedah@gmail.com')
validator('emmamail.com')
````
