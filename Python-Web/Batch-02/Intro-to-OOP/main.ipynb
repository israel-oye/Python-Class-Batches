```python
students = [
    {"id": 1, "name": "Alice", "scores": [10, 6, 7]},
    {"id": 2, "name": "Bankole", "scores": [8, 8, 8]},
    {"id": 3, "name": "Chidi", "scores": [6, 4, 4]},
    {"iD": 4, "name": "Desmond", "scoress": [6, 8, 4]},
]


def get_student(id_):
    for student in students:
        if student["id"] == id_:
            return student

get_student(id_=4)
```

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name # Attributes
        self.breed = breed

#       Dog.__init__()
bingo = Dog(name="Bingo") #constructor
jack = Dog(name='Jack')




print(bingo.name, jack.name)
```

```python
class Person:
    def __init__(self, fullname, gender):
        self.name = fullname
        self.sex = gender

person = Person(fullname='Gowon Jubee', gender='Male')

person.name
print(person.sex)
```
