2. *ATM Machine Simulation Using a While Loop*

*Objective:*  
Create a simple ATM machine program in Python using a while loop.  

*Requirements:*  
- Store a PIN in the program (e.g., `1234`).  
- Prompt the user to enter a PIN.  
- If the entered PIN is correct, display an ATM menu repeatedly using a while loop.  
- The menu should contain the following options:  
  1. Withdraw  
  2. View Balance  
  3. Exit  

*When the user selects:*  
- *Withdraw:* Display a message indicating that a withdrawal has been made.  
- *View Balance:* Display the current account balance.  
- *Exit:* Terminate the program.  

*Additional Rules:*  
- If the user enters an invalid menu option, display an appropriate error message and show the menu again.  
- If the PIN is incorrect, display an error message and do not show the ATM menu.

````python
import random

random_num = random.randint(1, 100)
attempts = 0

while True:
    guess = int(input("Enter guess: "))
    attempts += 1

    if guess == random_num:
        print(f"You guessed right. It took you {attempts} trials to get it")
        break
    elif guess < random_num:
        print(f"{guess} is too low.")
    elif guess > random_num:
        print(f"{guess} is too high")
    # print("Try again")
````

````python
# ATM 
pin = '1234'
balance = 300.5

user_pin = input("Enter PIN: ")

if user_pin == pin:
    while True:
        selected_menu = input("Menu\n\t1. Withdraw\n2. View balance\n3. Exit\n")

        if selected_menu == '1':
            amount_to_withdraw = float(input("How much to withdraw: "))
            if amount_to_withdraw > balance:
                print("INSUFFICIENT FUNDS...\n")
            else:
                balance -= amount_to_withdraw
        elif selected_menu == '2':
            print(f"Bal: ₦{balance:.2f}\n")
        elif selected_menu == '3':
            print("Exiting ATM. Thanks for banking with Nexus")
            break
        else:
            print("Invalid menu selected. Enter 1, 2, 3: ")
else:
    print("Invalid PIN.")
````
