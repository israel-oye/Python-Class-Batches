```python
# Database Connection with Python
import sqlite3


cnxn = sqlite3.connect('products.db')

cursor = cnxn.cursor()


# Tech gadgets table
# id, name, price

create_table_q = """
CREATE TABLE IF NOT EXISTS tech_gadgets (
id INTEGER PRIMARY KEY AUTOINCREMENT,
name TEXT, 
price REAL);
"""

cursor.execute(create_table_q)

cnxn.close()
```
