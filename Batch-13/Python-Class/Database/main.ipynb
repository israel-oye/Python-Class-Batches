```python
# Database Connection with Python
import sqlite3


cnxn = sqlite3.connect('products.db')

cursor = cnxn.cursor()

# Tech gadgets table
# id, name, price
def create_table():
    create_table_q = """
    CREATE TABLE IF NOT EXISTS tech_gadgets (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT, 
    price REAL);
    """

    cursor.execute(create_table_q)


def insert_data(product_name: str, price: float):
    try:
        cursor.execute(
            """
            INSERT INTO tech_gadgets (name, price)
            VALUES (?, ?);
            """,
            (product_name, price)
        )
    except sqlite3.Error as e:
        print("Something went wrong", e)
    else:
        cnxn.commit()

# create_table()
# insert_data('BT Speaker', price=449.99)


def fetch_data():
    cursor.execute(
        """
        SELECT * FROM tech_gadgets
        """
    )

    records = cursor.fetchall()
    print(records)


fetch_data()

# data_list = [
#     ("Phone", 5000.65),
#     ("Televison", 7999.99),
#     ("Decoder", 5400),
#     ("Wi-fi", 1250.45),
# ]

# for prd, price in data_list:
#     insert_data(prd, price)

# insert_data("BT Speaker", price=449.99)

cnxn.close()
```
