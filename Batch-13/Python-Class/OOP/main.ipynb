````python
class Rectangle:
    def __init__(self, length: float, width: float):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    # add a method for `perimeter`: 2 (l + b)
    def perimeter(self):
        return 2 * (self.length + self.width)

    def is_square(self):
        return self.length == self.width

    @staticmethod
    def describe():
        print("A rectangle is a 4-sided polygon where 2 sides are equal.")


rect1 = Rectangle(5, 3)
print(rect1.length)
print(rect1.width)

print(rect1.area())
print(rect1.perimeter())
print(rect1.is_square())

rect2 = Rectangle(10, 10)

print(rect2.area())
print(rect2.perimeter())
print(rect2.is_square())

Rectangle.describe()
````

````python
class Circle:
    def __init__(self, r: float):
        self.radius = r
        self.diameter = r * 2

    def area(self):
        from math import pi
        return pi * (self.radius ** 2)

    def circumference(self):
        return 2 * 3.142 * self.radius


c1 = Circle(r=14) # Circle.__init__()
print(c1.area())
print(c1.circumference())
print(c1.diameter)

c1.
````

````python
# Encapsulation: protecting data


class BankAccount:

    def __init__(self):
        self.__balance = 0.0

    # Getter
    def get_balance(self):
        return self.__balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, new_bal: float):
        self.__balance = new_bal


acc1 = BankAccount()

# acc1.__balance = 400
# print(acc1.__balance)
print(acc1.get_balance())
print(acc1.balance)
````

1. Create a `Vehicle` class with a method `start()`.
2. Create a `Car` class that inherits from `Vehicle` and adds a `fuel_type` attribute.
3. Use `super()` in the `Car` constructor to initialize the parent’s `brand` attribute.

````python
class Vehicle:
    def __init__(self, brand: str):
        self.brand = brand

    def start(self):
        return "Starting..."


class Car(Vehicle):
    def __init__(self, brand:str, fuel_type: str):
        super().__init__(brand=brand)
        self.fuel_type = fuel_type

    def start(self):
        print(f"{super().start()} with {self.fuel_type} engine")


car = Car('Toyota', 'Petrol')
car.start()
````
