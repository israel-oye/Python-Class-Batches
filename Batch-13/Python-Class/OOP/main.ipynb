```python
class Dog:
    # Constructor
    def __init__(self, dog_name: str, breed):
        # Attribute: a defining characterstic of an object
        self.name = dog_name
        self.breed = breed


#       Dog.__init__()
dog_1 = Dog(dog_name='Bingo', breed='Basenji')

dog_2 = Dog("Bullet", "rotweiler")


print(dog_1.name)
print(dog_1.breed, dog_2.breed)
```

```python
# Create a Student class with the following attributes:
#  - name
#  - gender
#  - grade
class Student:
    def __init__(self, name, gender, grade):
        self.name = name
        self.gender = gender
        self.grade = grade

student_1 = Student(
    name="Ajayi Somefun", 
    gender='Male',
    grade='12'
)

student_2 = Student(name="Bogunle Mabori", gender="Female", grade="10")

print(student_1.name.upper())
print(student_2.name)
```
