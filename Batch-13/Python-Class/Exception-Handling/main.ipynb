```python

# try:
#   something that can POTENTIALLY raise an error
# except ErrorName:
#   do sth if the error occurs
# else:
#   do sth if Error wasn't raised
# 

def add(x, y):
    try:
        result = x + y
    except TypeError:
        print("It seems you tried to add text and numbers")
    else:
        return result


add(4, 5)
print(add("4", 5))
```

```python
try:
    age = int(input("Enter age: "))
except ValueError as e:
    print(e)
    print("It seems you entered a wrong age format. Please enter age in digits ONLY")
else:
    yob = 2026 - age
    print(f"You were born  in {yob}")
finally:
    print("This will always run")
```

   Write a function `check_password(pw)` that raises a `ValueError` if the password is less than 6 characters. Catch and display a custom error message.


```python
def check_password(pw: str):
    if len(pw) < 6:
        raise ValueError('Password must be at least 6 characters')
    print("Password is good ✅")

try:
    check_password(pw=input("Enter password"))
except ValueError as e:
    print(e)
```
