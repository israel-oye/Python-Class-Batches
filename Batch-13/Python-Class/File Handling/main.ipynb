
### 📝 Assignment Task: Working with CSV Files in Python

You are provided with a CSV file containing names and ages of four students:

```
Name,Age
Ade,10
Bolu,11
Charles,12
Doreen,9
```

Your task is to **write a Python program** that opens this CSV file, reads the data, and computes the **average age** of the students listed.

---

### 🔍 What You Need to Do
1. **Open the CSV file** in Python.  
   - You can use the built-in `open()` function.  
   - Remember to open the file in **read mode** (`"r"`).

2. **Read the contents of the file.**  
   - Use `f.readlines()` to read all the lines into a list.  
   - Each line will be a string, for example: `"Ade,10\n"`.

3. **Extract the ages from each line.**  
   - Use the string method `.split(",")` to separate the name and age.  
   - The age will be the second item in the list after splitting.  
   - Don’t forget to convert the age from a string to an integer using `int()`.

4. **Compute the average age.**  
   - Add up all the ages.  
   - Divide the total by the number of students.

---

### 💡 Hints
- Skip the first line (`"Name,Age"`) because it is just the header.  
- Use a loop to go through each line after the header.  
- Keep track of the sum of ages and the count of students.  
- At the end, calculate the average age:  



---

### 🎯 Goal
By completing this task, you will practice:
- Reading files in Python  
- Working with strings and lists  
- Performing basic arithmetic operations  

The final output should display the **average age of the students**.

---


````python
with open('data.csv', 'r') as data_file:
    content = data_file.readlines()
    print(content)
    ages = []
    for line in content[1:]:
        parts = line.split(',')
        age = int(parts[1].strip())
        ages.append(age)

    average_age = sum(ages) / len(ages)
    print(ages)
    print(average_age)
````
