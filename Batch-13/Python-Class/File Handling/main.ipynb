```python
names = [
    ("Ade", 10),
    ("Bolu", 11),
    ("Charles", 12),
    ("Doreen", 9),
]

names = []
for data in names:
    print(data[0])

with open('data.csv', 'w') as f:
    # f.writelines(names)
    f.write("Name,Age")
    for pair in names:
        f.write(f"{pair[0]},{pair[1]}")
```

```python
names = {
    "Ade":  10,
    "Bolu": 11,
    "Charles": 12,
    "Doreen": 9,
}

with open('data-dict.csv', 'w') as f:
    f.write("Name,Age\n")
    for name, age in names.items():
        f.write(f"{name},{age}\n")
```
