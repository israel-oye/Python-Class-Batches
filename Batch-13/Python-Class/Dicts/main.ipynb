```python
students = {
    "Ade": [90, 85, 75],
    "Bilikis": [95, 80, 81],
    "Calistus": [65, 60, 85],
}

# Using a for loop print the name of each students and their average score
# e.g Name: Ade
#     Average: 83.33

# The highest score of all
# Who owns the highest score

for name, scores in students.items():
    # avg = sum(scores) / len(scores)
    # print(f"Name: {name}\tAverage: {avg}")
    student_total = 0
    for score in scores:
        student_total += score

    avg = student_total / len(scores)
    print(f"Name: {name}\tAverage: {avg:.2f}")


all_scores = []

for scores in students.values():
    all_scores.extend(scores)

print(all_scores)

max_score = max(all_scores)
# max_score = all_scores[0]

# for score in all_scores:
#     if score > max_score:
#         max_score = score

print(max_score)


for name, scores in students.items():
    if max_score in scores:
        print(f"{name} has the highest score {max_score}\tScores: {scores}")
```
