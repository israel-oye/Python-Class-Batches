````python
# Union of  set

set_a = {1, 2, 3, 5}
set_b = {2, 4, 6}

print(set_a | set_b)
print(set_a.union(set_b))
````

````python
# Difference

set_a = set('bobby')
print(set_a)

set_1 = {2, 3, 7, 11}
set_2 = {1, 3, 5, 7}

print(set_1 - set_2)
print(set_2 - set_1)

print(set_1.difference(set_2))
````

````python
# Symmetric difference

set_a = {1, 2, 3}
set_b = {3, 4, 5}

set_a.symmetric_difference(set_b)
````

````python
# Below is a record of attendances on two days.
mon = {"John", "Mary", "Steve"}
tues = {"Mary", "Steve", "Sam"}

# Write a program to print
# 1. Students who attended both days
print(f"Both days: {mon & tues}")

# 2. Students who attended only one day .
print(f"Only one day: {mon.symmetric_difference(tues)}")

# 3. Total unique students for both days .
print(f"All students for both days: {mon | tues}")
````

````python
# Below are records of registered and present students
registered = {"John", "Mary", "Steve", "Sam", "Lucy"}
present = {"Mary", "Steve", "Lucy"}

# Write a program to print the names of absent students
absent = registered - present

print(registered.symmetric_difference(present))
print(absent)
````

**Task:** Find which letters appear more than once in a person's full name.

```python
full_name = "Mississippi River"

print("All letters:", letters)
#print("Unique letters:", seen)
print("Duplicate letters:", duplicates)
```


````python
full_name = "Mississippi River"

letters = set()

for char in full_name:
    if char ! = ' ':
        letters.add(char.lower())
print(f"All letters: {letters}")

seen = set()
duplicates = set()

for char in full_name.lower():
    if char in seen:
        duplicates.add(char)
    else:
        seen.add(char)

print("Seen:", seen)
print("Duplicates:", duplicates)
````
