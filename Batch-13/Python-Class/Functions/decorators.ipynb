```python
# Decoprators

def func():
    return "This is a function"

print(func)

my_f = func
my_f()
```

```python
def executor(func):
    func()

def foo(x):
    print(x)

foo(x=1)
foo(1)

def greet(): 
    print("Hello world!")

def sy_hi():
    print("HIIIII!!")

executor(func=greet)
executor(sy_hi)
```

```python
def foo(param):
    print(param)

a = 10
b = 20

foo(param=a)
foo(b)
```

```python
def executor(func_1, func_2):
    print("This is executor running...")
    func_1()
    func_2()


def greet():
    print("Hello world!")

def say_hi():
    print("HIIIII!!")


executor(func_1=greet, func_2=say_hi)
```

```python
# x = 40

def outter():
    x = 4
    print('Inside function', x)

print('outside func', x)
outter()
```

```python
def outer():
    print("Outer is running...")

    def inner():
        print('Inner is running...')
    
    print(inner)

outer()
```
