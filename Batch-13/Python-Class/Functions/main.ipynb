```python
def get_initials(firstname: str, surname: str):
    """This function returns the initials of a name""" #docstring
    initials = f"{firstname[0]}.{surname[0]}".upper()
    return initials

# get_initials("Bolu", "ajayi") #-> "B.A"
# get_initials("Vijay", "Najeem")
# get_initials(surname='Gboluwaga', firstname='Ilesanmi')
fname = input("Enter your firstname: ")
lname = input("Enter last name: ")

print(f"Yout initials are: {get_initials(fname, lname)}")
```

```python
def analyze_nums(numbers: list):
    # TODO: assignment to include average in analysis
    
    analysis = {}
    even = []
    odd = []

    for n in numbers:
        if n % 2 == 0:
            even.append(n)
        else:
            odd.append(n)

    analysis['even'] = even
    analysis['odd'] = odd
    analysis['all'] = numbers

    return analysis
    


print(analyze_nums(numbers=[4, 11, 9, 17]))
print(analyze_nums([60, 22, 81, 99]))
```

```python
def adder(*nums):
    total = 0
    for n in nums:
        total += n

    print(total)


# adder('aw', 'rt', 'tt')
adder(5, 6, 7, 8, 9, 10)
adder(1, 2)
adder(50, 60, 40, 55, 666, 77, 88, 99)
```

```python
def describe_recipe(*ingredients, meal):
    print(f"Meal: {meal.title()}:\nIngredients:")

    for ingredient in ingredients:
        print(f"\t- {ingredient}")

describe_recipe('gari', 'water', 'salt', meal='Eba')

describe_recipe('rice', 'totmatoes', 'pepper', 'onions', 'seasoning', 'butter', meal='Jollof rice')
```

```python
def describe_recipe(meal, *ingredients):
    print(f"Meal: {meal.title()}:\nIngredients:")

    for ingredient in ingredients:
        print(f"\t- {ingredient}")

describe_recipe("gari", "water", "salt", "Eba")
describe_recipe("boiled water")
```
