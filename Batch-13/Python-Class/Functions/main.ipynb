*Exercise 3: Classroom Statistics*  

Create a function called `class_statistics()` that accepts any number of student scores and displays:  
- The highest score  
- The lowest score  
- The average score  
- The total number of scores  

*Example*  
`class_statistics(75, 80, 92, 67, 88)`  

Output:  
`Highest Score: 92`  
`Lowest Score: 67`  
`Average Score: 80.4`  
`Total Scores: 5`

````python
def class_statistics(*scores):
    print(f"Highest: {max(scores)}")
    print(f"Highest: {min(scores)}")
    print(f"Average: {sum(scores) / len(scores):.2f}")
    print(f"Total scores: {len(scores)}")


class_statistics(80, 85, 77, 91, 65)
````

Create a function called `report_card()` that accepts a student's subjects and their corresponding scores as keyword arguments. The function should print out 
- the average of the student's scores
- the highest of their scores
- the lowest of their scores
- the number of subjects

The output should look like the following example: 
``` python
report_card(English=71, Maths=81, Bio=95)
# REPORT CARD
# -------------
# English: 71
# Maths: 81
# Bio: 95

# Average: 42.3
# Highest: 95
# Lowest: 71
# Subjects registered: 3
```

````python
def report_card(**score_data):
    print("REPORT CARD\n----------")
    for course, score in score_data.items():
        print(f"{course.title()}: {score}")

    all_scores = score_data.values()
    registered_subjects_count = len(all_scores)

    print(f"\nRegistered subjects: {registered_subjects_count}")
    print(f"Highest score: {max(all_scores)}")
    print(f"Lowest score: {min(all_scores)}")
    print(f"Average score: {sum(all_scores) / len(all_scores)}")
    


report_card(English=71, Maths=81, Bio=95, Phy=90)
print()
report_card(civic=95, phe=87, cca=88, phonics=75)
````
