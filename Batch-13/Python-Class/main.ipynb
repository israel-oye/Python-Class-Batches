```python
# Arsenal has 54 points currently at M37
# yOU ARE TO CREATE A PROGRAM THAT COMPUTES THE POINTS won in a league match
# Win: 3 pts
# Draw: 1 pt
# Loss: 0pt
# An official enters the outcome of an Arsenal game 'win', 'draw' or 'loss'
# Depending on the match outcome, compute the points that Arsenal has
points = 54

outcome = input("Match outcome: ")

if outcome == 'win':
    points += 3
elif outcome == 'draw':
    points += 1
elif outcome == 'loss':
    points += 0
else:
    print("Invalid outcome {outcome}\tValid outcomes are 'win', 'draw' or 'loss' ")

print(f"Arsenal points after match {outcome} = {points} pts")
```
