```python
# json.dumps({'a', 'b'})


class User:
    def __init__(self, name: str, gender: str, email: str):
        self.name = name
        self.gender = gender
        self.email = email
    
    def to_json(self):
        import json
        return json.dumps({
            'name': self.name, 
            'gender': self.gender,
            'email': self.email
            })

user = User(name='Ayo', gender='Male', email='ayo@mail.com')
user.to_json()

user.__dict__
```
