```python
# json.dumps({'a', 'b'})


class User:
    def __init__(self, name: str, gender: str, email: str):
        self.name = name
        self.gender = gender
        self.email = email
    
    def to_json(self):
        import json
        return json.dumps({
            'name': self.name, 
            'gender': self.gender,
            'email': self.email
            })

user = User(name='Ayo', gender='Male', email='ayo@mail.com')
user.to_json()

user.__dict__
```

```python
import json

students = [
    {"name": "Ali", "grade": 85},
    {"name": "Zara", "grade": 92},
    {"name": "John", "grade": 78},
]

# Serialize this to a students.json file
with open('students.json', 'w') as f:
    json.dump(students, f, indent=4)
```

```python
# DESERIALIZING FROM JSON FILE

import json

with open('user.json', 'r') as f:
    obj = json.load(f)
    print(obj)
    print(type(obj))
```

```python
# Pickling: is Python-specific serialization
import pickle

students = [
    {"name": "Ali", "grade": 85},
    {"name": "Zara", "grade": 92},
    {"name": "John", "grade": 78},
]

with open('students.pkl', 'wb') as f:
    pickle.dump(students, f)
```

```python
import pickle

with open('students.pkl', 'rb') as f:
    obj = pickle.load(f)
    print(obj)
    print(obj[0])
```
