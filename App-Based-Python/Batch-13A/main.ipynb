 Number Guessing Game**

 *Concepts: Loops, conditionals, user input*

* The program selects a **random number between 1 and 100**.
* The user keeps guessing until they get it right.
* The program gives hints ("too high" or "too low").

```python
import random

attempts = 0
random_num = random.randint(1, 100)

while True:
    guess = int(input("Guess the number: "))
    attempts += 1

    if not 0 <= guess <= 100:
        print("Guess is out of range")

    if guess < random_num:
        print(f"You guessed too low: {guess}")
    elif guess > random_num:
        print(f"You guessed too high {guess}")
    else:
        # correction
        print(f"Correct! It took you {attempts} trials to get it. {random_num}")
        break
```

```python
def transform_numbers(numbers):
    result = []
    for num in numbers:
        doubled = num * 2
        if doubled <= 10:
            result.append(doubled)
    return result


# Test cases that should work:
print(transform_numbers([3, 5, 8, 10]))  # Expected: [16, 20]
print(transform_numbers([1, 2, 3]))  # Expected: []
print(transform_numbers([6, 7, 8]))  # Expected: [12, 14, 16]
```

```python
def process_strings(strings):
    cleaned = []
    for word in strings:
        upper_word = word.upper()
        if len(word) >= 3:
            if upper_word not in cleaned:
                cleaned.append(upper_word)
    return cleaned


# Test cases that should work:
print(process_strings(["hello", "world", "hi", "hello"]))  # Expected: ['HELLO', 'WORLD']
print(process_strings(["a", "bb", "ccc", "dd"]))  # Expected: ['CCC']
print(process_strings(["apple", "Apple", "APPLE"]))
```
