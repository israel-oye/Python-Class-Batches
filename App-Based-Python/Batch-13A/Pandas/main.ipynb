```python
import pandas as pd

df = pd.read_csv('students.csv')

df['Hobbies'] = df['Hobbies'].str.upper()
df['Gender'] = df['Gender'].str.replace('Male', 'M')
df["Gender"] = df["Gender"].str.replace("Female", "F")

# display(df)
df.isnull().sum()
df.duplicated()
```

```python
display(df.head(3))
print("\n\n\n")
display(df.tail(4))
```

```python
dancers = df[df['Hobbies'] == 'Dancing']

display(dancers)
```

```python
df['Average'] = df[['C++ Marks', 'Java Marks', 'Python Marks']].mean(axis=1)

df

sorted_names_df = df.sort_values(by=['Name'], ascending=[True])

display(sorted_names_df)

print("\n\n\n\n")


sorted_averages_df = df.sort_values(by=["Average"], ascending=[False])
display(sorted_averages_df)
```
