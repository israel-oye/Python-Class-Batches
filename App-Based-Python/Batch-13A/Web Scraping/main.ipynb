```python
from bs4 import BeautifulSoup

html = """
<html>
    <body>
        <h1>Python Course</h1>
        <h1>Springboot</h1>
    </body>
</html>
"""

html_doc = """
<html><head><title>The Dormouse's story</title></head>
<body>
<p class="title"><b>The Dormouse's story</b></p>

<p class="story">Once upon a time there were three little sisters; and their names were
<a href="http://example.com/elsie" class="sister" id="link1">Elsie</a>,
<a href="http://example.com/lacie" class="sister" id="link2">Lacie</a> and
<a href="http://example.com/tillie" class="sister" id="link3">Tillie</a>;
and they lived at the bottom of a well.</p>

<p class="story">...</p>
"""

soup = BeautifulSoup(html, "html.parser")
soup_2 = BeautifulSoup(html_doc, "html.parser")
# print(soup.prettify())
# Finding a single element
heading = soup.find('h1')
print(heading.text)
print(soup.find_all('h1'))

print(soup_2.head.title)

for link in soup_2.find_all('a'):
    print(link["href"])
```

```python
from bs4 import BeautifulSoup


with open('index.html', 'r') as file:
    html_content = file.read()

soup = BeautifulSoup(html_content, "html.parser")

p_elem = soup.find('p')

words = p_elem.text.split()
print(f"{len(words)} words in the first paragraph")
```

```python
from bs4 import BeautifulSoup

with open("index.html", "r") as file:
    html_content = file.read()

soup = BeautifulSoup(html_content, "html.parser")

sel_btn = soup.find(id='sel-btn')
print(sel_btn)

print(soup.find(class_='text'))

print(soup.select('h1.heading'))
print(soup.select('#click-btn'))
print(soup.select_one('button'))
```
