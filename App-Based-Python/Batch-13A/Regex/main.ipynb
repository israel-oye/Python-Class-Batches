```python
import re

text = "My num is 080123456"

match = re.search(r"[a-z]", text)
print(match.group())

# 0 OR MORE
print(re.findall(r"[a-z]*", text))

# 1 or more
print(re.findall(r"[a-z]+", text))

# 0 OR 1 cases
print(re.findall(r"[A-Z]?", text))

(re.findall(r"[0-9]+", text))
```

```python
import re

text = "My num is 080123456 J500 Emma123"

print(re.findall(r"[0-9]{6,11}", text))

print(re.findall(r"[a-zA-Z]+[0-9]{3}", text))

# r"[a-zA-Z0-9_.+-]"

text = "2026 is a great year"

re.findall(r"[0-9]{4}", text)
re.findall(r"\d{4}", text)
```

```python
import re

string = "Room 12, Floor 3, Block 402, Unit 7"

# use `re.findall()` to extract only the numbers that are 2 digits or more. 
# Expected: `['12', '402']`.

re.findall(r"\d{2,}", string)
```

```python
import re

text = "Errors: ERROR_404, ERROR_500, ERROR_403"

re.findall(r"ERROR_\d{3}", text)


re.findall(r"^class", "hello class")

re.findall(r"[^aeiou\s]", "hello world")

re.findall(r"[a|b]", "hoola baloo baloon")

re.findall(r"^0(80|81|70)", "08123456")
```
