```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.cluster import KMeans
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('monthylysavings.csv')

# display(df)
x = df.iloc[:, [2, 3]]
display(x)

ms = MinMaxScaler()
x = ms.fit_transform(x)
x
kmeans = KMeans(n_clusters=3)
y_predict = kmeans.fit_predict(x)

display(x, y_predict)

display(x[y_predict == 0, 0])

plt.scatter(x[y_predict == 0, 0], x[y_predict == 0, 1], s=100, c="r", label="Cluster 0")
plt.scatter(x[y_predict == 1, 0], x[y_predict == 1, 1], s=100, c="g", label="Cluster 1")
plt.scatter(x[y_predict == 2, 0], x[y_predict == 2, 1], s=100, c="b", label="Cluster 2")


plt.xlabel("Monthly Income ($)")
plt.ylabel("Savings score")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np

# Load dataset
df = pd.read_csv("age_heights.csv")

# Features (Age) and target (Height)
X = df[["Age"]]  # Independent variable
y = df["Height_cm"]  # Dependent variable

# Train linear regression model
model = LinearRegression()
model.fit(X, y)

# Predict values
ages = np.linspace(df["Age"].min(), df["Age"].max(), 100).reshape(-1, 1)
predicted_heights = model.predict(ages)

# Plot data points
plt.scatter(
    df["Age"],
    df["Height_cm"],
    c=df["Gender"].map({"Male": "blue", "Female": "red"}),
    label="Data",
)

# Plot regression line
plt.plot(ages, predicted_heights, color="green", linewidth=2, label="Regression Line")

# Labels and legend
plt.xlabel("Age (years)")
plt.ylabel("Height (cm)")
plt.title("Relationship between Age and Height")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.model_selection import train_test_split


df = pd.read_csv('icecream_sales.csv')

X = df[['Temperature (Celsius)']]
y = df[['Ice-cream Sales (kg)']]
df

x_train, x_test, y_train, y_test =  train_test_split(X, y, test_size=0.3, random_state=2)

model = LinearRegression()
model.fit(x_train, y_train)

y_prediction = model.predict(x_test)
x_test, y_prediction

comparism_results = pd.DataFrame({'Actual value': y_test.squeeze(), 'Predicted value': y_prediction.flatten()})
comparism_results


# Actual values in dataset that model wasn't trained on
plt.scatter(x_test, y_test, s=40, c='green')
plt.scatter(x_test, y_prediction, s=40, c='blue')
plt.plot(x_test, y_prediction, c='red', label='Regression line')

plt.xlabel('Temperature (Celsius)')
plt.ylabel('Ice-cream sales (kg)')
plt.legend()
plt.show()

# RSQ - R-Squared value of pur model
model.score(X, y)

meanAbsErr = mean_absolute_error(y_test, y_prediction)
meanAbsErr

meanSqrErr =  mean_squared_error(y_test, y_prediction)
meanSqrErr

#TODO: COmmit folder to github
```
