```python
import tkinter as tk

window = tk.Tk()
window.title("My App")

# window.geometry("300x200")
window.minsize(width=300, height=250)



# Label Widget
label = tk.Label(window, text='Hello Tkinter')
label.config(font=("Arial", 18, "bold"))
label.pack(side='bottom')

# Entry widget
text_input = tk.Entry(text=".")
text_input.pack()

# 
result_lbl = tk.Label()

def on_click():
    user_input = float(text_input.get())
    km = user_input / 1000
    result_lbl.config(text=f"{user_input}m => {km}km")
    text_input.delete(0, tk.END)
    

# Button widget
btn = tk.Button(text='Click', command=on_click)
btn.pack()
result_lbl.pack()


window.mainloop()
```

```python
import tkinter as tk


window = tk.Tk()
window.title('Login Form')
window.geometry('350x250')


username_store = 'user'
password_store = 'password'

username_lbl = tk.Label(window, text="Username")
username_entry = tk.Entry()


password_lbl = tk.Label(window, text="Password")
password_entry = tk.Entry()

def login():
    username = username_entry.get()
    password = password_entry.get()

    if username == username_store and password == password_store:
        message_lbl.config(text="Login Success...✅")
    else:
        message_lbl.config(text="Invalid credentials...❌")


login_btn = tk.Button(text='Login', command=login)
message_lbl = tk.Label(font=('Arial', 14, 'bold'))


username_lbl.pack()
username_entry.pack()
password_lbl.pack()
password_entry.pack()

login_btn.pack()
message_lbl.pack(pady=16)

window.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.geometry('300x200')

def on_click():
    messagebox.showinfo(title='Message', message='Button was clicked...')
    user_choice = messagebox.askokcancel(title='Confirm Action', message='re you sure about this action?')
    print(user_choice)

btn = tk.Button(text='Click', command=on_click)
btn.pack()

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.geometry("300x200")

gender = tk.StringVar()
gender.set('F')

male_btn = tk.Radiobutton(text='Male', variable=gender, value='M', command=lambda: print('Male selected', gender.get()))
female_btn = tk.Radiobutton(
    text="Female",
    variable=gender,
    value="F",
    command=lambda: print("Female selected", gender.get()),
)

bool_var = tk.BooleanVar(value=True)

check_btn = tk.Checkbutton(text='Check', variable=bool_var)
check_btn.pack(pady=24)

male_btn.pack()
female_btn.pack()
root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.geometry("300x200")
root.title('Toggler')
root.config(bg='snow')

is_dark = tk.BooleanVar(value=False)

def on_toggle():
    root.config(bg='grey65') if is_dark.get() else root.config(bg='snow')
    # if is_dark.get() == True:
    #     root.config(bg='grey65')
    # else:
    #     root.config(bg="snow")

dark_mode_btn = tk.Checkbutton(text='Dark mode', variable=is_dark, command=on_toggle)
dark_mode_btn = tk.Checkbutton(
    text="Dark mode",
    variable=is_dark,
    command=lambda: (
        root.config(bg="grey65") if is_dark.get() else root.config(bg="snow")
    ),
)

dark_mode_btn.pack(pady=24)

# STOCK
root.mainloop()
```
