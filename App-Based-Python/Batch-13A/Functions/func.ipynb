Write a function that accepts a list of items a person wants to buy at the grocery store. The function should have one parameter that collects as many items as the function call provides, and it should print a summary of the shopping list. Call the function three times, using a different number of arguments each time.

Your function should work for calls like these:
```
make_shopping_list('milk')
make_shopping_list('eggs', 'bread', 'butter')
make_shopping_list('chicken', 'rice', 'broccoli', 'garlic', 'soy sauce', 'ginger')


````python
def make_shopping_list(*shopping_list):
    print("Shopping list: ")
    for item in shopping_list:
        print(f"\t-{item}")

make_shopping_list()
make_shopping_list("milk")
make_shopping_list("eggs", "bread", "butter")
make_shopping_list("chicken", "rice", "broccoli", "garlic", "soy sauce", "ginger")
````

````python
# Arbitrary Keyword arguments

def bio_maker(name, age, **details):
    # print(details)
    print(f"Name: {name}\nAge:{age}")
    for key in details:
        print(f"{key}: {details[key]}")

bio_maker(name='Ade', age=21)
bio_maker('Bolu', age=20, height=1.88, nationality='Nigeria')
bio_maker('Raza', 22, hobbies=['football', 'dancing'], middle_name='Resha')
````

````python
def factorial(n: int):
    if n == 1:
        return 1
    return n * factorial(n- 1)

factorial(3)
````

````python
# Higher Order Functions: a function that can take a func as a parameter

def detailer(func):
    print("Before function call")
    func()
    print("After function call")

def greet():
    print("Hello world!")

def speak():
    print('lorem ipsum')

detailer(func=greet)
detailer(speak)
````

````python
# Map(a function, a collection)
def to_upper(string: str):
    return string.upper()

names = ['Zoe', 'Yu', 'Xiao', 'Wong']

# for name in names:
#     print(to_upper(name))
print(list(map(to_upper, names)))
print(list(map(lambda txt: txt.upper(), names)))
````

````python
# Filter: HOF
def is_even(n: int):
    return n % 2 == 0

print(list(filter(is_even, range(11))))

# Modify this code to filter out multiples of 3 (use lambda function)

list(filter(lambda n: n % 3 == 0, range(11)))
````
