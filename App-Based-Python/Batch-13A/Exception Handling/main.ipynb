```python
# num_1 = int(input("Enter first number: "))
# num_2 = int(input("Enter second number: "))


def divide(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print(f"You can't divide {x} by {y}")
    except TypeError:
        print("Can only divide numbers")
    # except Exception:
    else:
        return result


# print(divide(x=num_1, y=num_2))
divide('10', 2)
```

```python
try:
    age = int(input("Enter age: "))
except ValueError:
    print("Please enter your age in digits alone")
else:
    print(f"Age: {age}")
finally:
    print("This runs always")

```

```python
try:
    f = open('heroes.txt', 'r')
    content = f.read()
    print(content)
except FileNotFoundError as exc:
    print("File at path doesn't exist")
finally:
    f.close()
```

```python
names = [
    'Abeeb',
    'Bolu',
    'Charles'
]

position = int(input("S/N: "))

try:
    name = names[position - 1]
except IndexError as e:
    print("Invalid S/N", e)
else:
    print(name)
```
