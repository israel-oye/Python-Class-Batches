```python
with open('new.txt', mode='w') as f:
    user_entry = input("Enter what you want to save")
    f.write(user_entry)
    
```

```python
with open('students.txt', 'w') as f:
    # f.write("1. Ada\n2. Bimbo\n3. Charles")
    students = ['Afellay', 'Barbosa', 'Cucho']
    for student in students:
        f.write(student + '\n')
```

```python
# Appending mode

with open('students.txt', mode='a') as f:
    f.write('Domingo\n')
```

```python
# Reading mode

with open('students.txt', 'r') as f:
    # print(f.readlines())
    print(f.readlines())
    # print(f.read())
```

```python
with open('poem.txt', 'r') as f:
    lines = f.readlines()
    print(lines)
    print(len(lines))
```
