```python
root = tk.Tk()
root.title("Hobbies")

tframe = tk.Frame(root)
tframe.pack()

bframe = tk.Frame(root)
bframe.pack(side=tk.BOTTOM)

lframe = tk.Frame(bframe)
lframe.pack(side=tk.LEFT)

rframe = tk.Frame(bframe)
rframe.pack()

read = tk.IntVar()
song = tk.IntVar()


def show_hobbies():
    if read.get() and song.get():
        message = "Reading books and\nListening to songs"
    elif song.get():
        message = "Listening to songs"
    elif read.get():
        message = "Reading books"
    else:
        message = "No hobby selected"

    dmessage = tk.Label(
        rframe,
        text=message,
        bg="blue",
        fg="white"
    )
    dmessage.pack()


tk.Checkbutton(
    tframe,
    text="Reading books",
    variable=read,
    onvalue=1,
    offvalue=0
).grid(column=0, row=0, sticky=tk.W)

tk.Checkbutton(
    tframe,
    text="Listening to songs",
    variable=song,
    onvalue=1,
    offvalue=0
).grid(column=1, row=0, sticky=tk.W)

button1 = tk.Button(
    lframe,
    text="Click to display selected hobbies.",
    bg="red",
    padx=5,
    pady=5,
    command=show_hobbies
)
button1.pack()

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.title("My App")
root.geometry("400x300")
root.minsize(width=400, height=300)

count = 0

label = tk.Label(root, text=f"Button was clicked {count} times")
label.config(
    font=("Aptos", 16, "bold"),
)
label.pack()
# Layout Managers


def foo():
    global count
    print("Button was clicked")
    count += 1
    label.config(text=f"Button was clicked {count} times")


btn = tk.Button(text="Click", bg="deep sky blue", fg="white", command=foo)
btn.pack()

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.title("Dark Mode")
root.minsize(width=300, height=250)
root.config(bg='snow')

def toggle():
    if root.cget('bg') == 'snow':
        root.config(bg='grey37')
    else:
        root.config(bg='snow')


btn = tk.Button(root, text='Switch theme', command=toggle)
btn.pack()

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.title("Entry")
root.minsize(width=300, height=250)


input_box = tk.Entry()
input_box.pack()

display_lbl = tk.Label()

def input_handler():
    user_entry = input_box.get()
    input_box.delete(0, tk.END)
    display_lbl.config(text=user_entry)

btn = tk.Button(text='Get', command=input_handler)
btn.pack()


display_lbl.pack()

root.mainloop()
```

create a registration form app that takes 
- the user's name 
- the user's email and
- the user's date of birth 
and save those details in a CSV file.

```python
import os
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.title("Registration Form")
root.minsize(width=300, height=250)

frame = tk.Frame(root)


username_lbl = tk.Label(frame, text='Username')
username_entry = tk.Entry(frame)

email_lbl = tk.Label(frame, text="Email")
email_entry = tk.Entry(frame)

dob_lbl = tk.Label(frame, text="Date of Birth")
dob_entry = tk.Entry(frame)

gender_row = tk.Frame(frame)
gender_var = tk.StringVar(value="F")

male_btn = tk.Radiobutton(gender_row, text="Male", value="M", variable=gender_var)
female_btn = tk.Radiobutton(gender_row, text="Female", value="F", variable=gender_var)


def on_submit():
    username = username_entry.get().strip()
    email = email_entry.get().strip()
    dob = dob_entry.get().strip()
    gender = gender_var.get()

    #Resetting the inputs
    username_entry.delete(0, tk.END)
    email_entry.delete(0, tk.END)
    dob_entry.delete(0, tk.END)
    gender_var.set("")

    # Validation check
    if not username or not email or not dob:
        messagebox.showerror(title="Invalid input", message="Please fill in all input boxes")
        return
    
    save_user(
        username,
        email,
        dob,
        gender
    )
    messagebox.showinfo(title='Insert Sucess', message='Added successfully')


def create_file():
    if not os.path.exists('users.csv'):
        with open ('users.csv', 'w') as f:
            f.write("Username,Email,DOB,Gender\n")
    

def save_user(username: str, email: str, dob: str, gender: str):
    with open('users.csv', 'a') as f:
        f.write(f"{username},{email},{dob},{gender}\n")


submit_btn = tk.Button(frame, text="Submit", command=on_submit)


def display():
    username_lbl.pack()
    username_entry.pack()

    email_lbl.pack()
    email_entry.pack()

    dob_lbl.pack()
    dob_entry.pack()

    male_btn.pack(side='left')
    female_btn.pack(side='right')
    
    gender_row.pack()
    submit_btn.pack()

    frame.pack()

display()
create_file()

root.mainloop()
```
