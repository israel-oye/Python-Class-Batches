```python
# Data cleaning

import pandas as pd

df = pd.read_csv("student_success.csv")

# print(df.shape)

# df.info()

display(df)
display(df["Hours_attended"].sum())
display(df.isnull()["Gender"].sum())
display(df.isnull().sum())
# df.duplicated()

df["Success_exam"] = df["Success_exam"].str.replace("Yes", "Y")
df["Success_exam"] = df["Success_exam"].str.replace("No", "N")

df["Success_exam"] = df["Success_exam"].fillna("N")

df

avg_hours_attended = df["Hours_attended"].mean()

df["Hours_attended"] = df["Hours_attended"].fillna(avg_hours_attended)
```

```python
df.duplicated()

df = df.drop_duplicates()

df.describe()

df.values

df.Courses.values
```
