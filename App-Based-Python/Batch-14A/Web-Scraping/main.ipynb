```python
from bs4 import BeautifulSoup

html_text = """
<html>
    <body>
        <h1>Python Course</h1>
        <h1>Springboot</h1>
        <div>
            <h1>Java Course</h1>
            <h1 class='w'>Winter</h1>
        </div>
        <p>lorem ipsum dolor sit met</p>
    </body>
</html>
"""

soup = BeautifulSoup(html_text, "html.parser")

print(soup.find('h2'))
soup.find_all('h1')

print(soup.p)
soup.div.find_all('h1')[1]['class']
```

```python
web_page = """
<a href="https://example.com/alice">Alice</a>
<a href="https://example.com/beatrice">Beatrice</a>
<a href="https://example.com/celice">Celice</a>
"""

soup = BeautifulSoup(web_page, "html.parser")
first_link = soup.find('a')
first_link['href']


links = []
for a in soup.find_all('a'):
    links.append(a['href'])


links
```

```python
import requests

resp = requests.get("http://127.0.0.1:5000/products")
# print(resp.text)

soup = BeautifulSoup(resp.text, 'html.parser')

products = []

for a in soup.find_all('a'):
    products.append(a.text.strip())

print(products)
```
