```python
import requests
from bs4 import BeautifulSoup
from openpyxl import Workbook
from openpyxl.styles import Font


URL = "https://books.toscrape.com/catalogue/page-2.html"
wb = Workbook()
ws = wb.active



resp = requests.get(url=URL)

soup = BeautifulSoup(resp.text, "html.parser")

book_titles = []
books_div = soup.section.find_all('div')[1]


books_div.find_all('li')[0].find_all('a')[1]

# print(books_div.find_all("li"))
for li in books_div.find_all('li')[:20]:
    title_anchor = li.find_all('a')[1]
    book_titles.append(title_anchor.text.removesuffix("...").strip())

book_titles

ws['A1'] = 'Book Title'
ws['A1'].font = Font(size=14, bold=True)

for idx, book in enumerate(book_titles):
    ws[f"A{idx + 2}"] = book


wb.save('book-titles.xlsx')
```
