```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import metrics
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression


df = pd.read_csv('sales_icecream.csv')
model = LinearRegression()


x = df[["Temperature (Celsius)"]]
y = df["Ice-cream Sales (kg)"]

# display(df)

x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.7, random_state=40)

# display(x_train, y_train, x_test, y_test)
# Training the model
model.fit(x_train, y_train)
# x_test = [40, 65, 43]
y_pred = model.predict(x_test)

# print(x_test, y_pred)

result_df = pd.DataFrame(data={
    'Temperature': x_test.to_numpy().flatten(),
    'Actual Sales': y_test,
    'Predicted Sales': y_pred
})

result_df
# df["Temperature (Celsius)"]

plt.scatter(x_test, y_test, color='green')
plt.scatter(x_test, y_pred, color='red')
plt.plot(x_test, y_pred, 'b')
plt.xlabel("Temperature (C)")
plt.ylabel("Ice-cream Sales (kg)")

plt.show()

rSqM = model.score(x, y)
rSqM

mAbsErr = metrics.mean_absolute_error(y_test, y_pred)
mSqErr = metrics.mean_squared_error(y_test, y_pred)

print(rSqM, mAbsErr, mSqErr)
```
