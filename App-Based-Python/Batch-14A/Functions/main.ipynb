```python
def describe_pet(animal, name):
    print(f"I have a/an {animal} named {name}")

# Positional arguments
describe_pet('dog', 'Bingo')

describe_pet('kathryn', 'cat')

# Keyword arguments
describe_pet(name='Messi', animal='goat')


describe_pet('dog', name='Doug')
```

Write a function called city_country() that takes in the name
of a city and its country. 

The function should return a string formatted like this:
"Santiago, Chile"

Call your function with at least three city-country pairs, and print the value that’s returned.




```python
def city_country(city, country):
    return f"{city}, {country}"


print(city_country('Calabar', 'Nigeria'))

print(city_country(country='USA', city='New York'))
```
