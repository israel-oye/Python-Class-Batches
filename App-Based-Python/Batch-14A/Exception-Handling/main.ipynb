```python
import math


def division(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print("Sorry can't divide by 0")
        return math.inf
    except TypeError:
        print("Invalid data types. Can only divide numbers")
    except Exception:
        print("An error occured...")
    else:
        return result
    


# x = int(input("Enter first number: "))
# y = int(input("Enter second number: "))


# print(f"{x} ⁒ {y} = {division(x, y)}")
# int('ten')
[1, 2, 3][10]
division(5 , '2')
```

```python
try:
    result = int('70') / 0
except ZeroDivisionError:
    print("Zero division")
except ValueError:
    print("Invalid")
else:
    print("No errors, Result: ", result)
finally:
    print("This always runs")
```

```python
try:
    f = open('file.txt', 'rw')
except FileNotFoundError:
    print("File not found")
except ValueError:
    pass
finally:
    f.close()
```

```python
def age_checker():
    age = int(input("Enter age: "))
    if age < 0:
        raise ValueError("Age cannot be negative")
    else:
        return age

try:
    age = age_checker()
except ValueError as e:
    print(e)
else:
    print(age)
```
