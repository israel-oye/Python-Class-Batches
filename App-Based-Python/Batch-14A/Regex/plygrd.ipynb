```python
import re

text = "The prices are 500, 1200 and 75."

re.findall(r'\d+', text)
```

```python
text = """
python
pyramid
pycharm
Java
"""

re.findall(r'^py\w+', text, re.MULTILINE)
```

```python
text = """
08012345678
08123456789
07012345678
09012345678
"""

re.findall(r"0\d{10}", text, re.MULTILINE)
```

```python
text = """
08012345678
08123456789
07012345678
09012345678
"""

re.search(r"^0(70|80|90|81)\d{8}$", text, re.MULTILINE).group(0)
```
