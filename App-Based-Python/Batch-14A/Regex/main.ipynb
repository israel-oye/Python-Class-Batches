```python
import re

text = "I enjoy coding & cooking with Anjola at 5pm."

re.findall(r'[.]', text)
```

.       - Any Character Except New Line
\d      - Digit (0-9)
\D      - Not a Digit (0-9)
\w      - Word Character (a-z, A-Z, 0-9, _)
\W      - Not a Word Character
\s      - Whitespace (space, tab, newline)
\S      - Not Whitespace (space, tab, newline)

Quantifiers:
*       - 0 or More
+       - 1 or More
?       - 0 or One
{3}     - Exact Number
{3,4}   - Range of Numbers (Minimum, Maximum)


```python
re.findall(r"\d", "234 is Nigeria's code")
re.findall(r"\d", "234 is Nigeria's code")
re.findall(r"\w", "234 is Nigeria's code")
re.findall(r"\w", "234 is Nigeria's code")
re.findall(r"\s", "234 is\t Nigeria's code\n")
re.findall(r"\S", "234 is\t Nigeria's code\n")
re.findall(r"\d\s", "234 is\t Nigeria's code\n")
re.findall(r"\w{3}", "2345678 is\t Nigeria's code\n")


re.findall(r"\d*", "code is 0024")
re.findall(r"\d+", "code is 0024 2")
re.findall(r"\d?", "code is 0024 2")

```

```python
import re

with open('prose.txt', 'r') as f:
    words = f.read()

    word_list = re.findall(r'\b\w+\b', words)
    # word_list = [word.lower() for word in word_list]

    word_list = list(map(lambda w: w.lower(), word_list))

word_count = {}

for w in word_list:
    if w in word_count:
        word_count[w] = word_count[w] + 1
    else:
        word_count[w] = 1

# Try to implement a sorted dictionary where the earliest
# keys have the highest values
```

```python
import re

with open('prose.txt', 'r') as f:
    content = f.read()

    sentences = re.split(r'[.!?]', content)
    for idx, s in enumerate(sentences):
        print(f"{idx+1}. {s}")
```
