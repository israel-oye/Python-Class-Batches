**Time of Day Greeter**
   Input the current hour (0–23). Respond with:

* “Good morning” (5–11)
* “Good afternoon” (12–16)
* “Good evening” (17–20)
* “Good night” (21–4)

```python
hour_of_day = int(input("Enter hour of day: "))

if 5 <= hour_of_day <= 11:
    print("Good morning...")
elif 12 <= hour_of_day <= 16:
    print("Good afternoon")
elif 17 <= hour_of_day <= 20:
    print("Good evening")
elif (21 <= hour_of_day <= 23) or (0 <= hour_of_day <= 4):
    print("Good night...")
else:
    print("Invalid hour (0-23)")
```
