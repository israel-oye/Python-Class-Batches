````python
# For loops

# For item in collection:
#     do sth with the item

for i in range(6):
    print(i)

for friend in ['Beks', 'Chuks', 'Desmond']:
    print(friend)

for letter in set("bcdef"):
    print(letter)


# Using  for-loop, print out ll MULTIPLES OF 5 BETWEEN 50-100

for j in range(50, 100, 5):
    print(j)
````

````python
numbers = range(1, 6)

doubled_nums = []
squared_nums = []
cube_nums = []

for num in numbers:
    doubled_nums.append(num * 2)
    squared_nums.append(num ** 2)
    cube_nums.append(num ** 3)

print(doubled_nums, squared_nums, cube_nums, sep="\n")
````

````python
# Using a for-loop,
# create a list of all squares between 1 and 10
squares = []

for i in range(1, 11):
    squares.append(i ** 2)

squares
````

````python
odd_nums = []

for i in range(11):
    if i % 2 == 1:
        odd_nums.append(i)

odd_nums
````

````python
temperatures = [15, 22, 30, 35, 40, 5]

# - Create a new list that contains only the temperatures above 20 degrees Celsius
# - Print the list
temps_above_20 = []

for temp in temperatures:
    if temp > 20:
        temps_above_20.append(temp)

temps_above_20
````

````python
words = ["hello", "world", "python", "rocks"]

# - Create a new list that contains each word reversed.
# - Print the new list.

reversed_words = []

for word in words:
    reversed_words.append(word[::-1])

reversed_words
````

````python
words = ["apple", "banana", "kiwi", "cherry", "fig"]

words_fewer_than_4 = []

for word in words:
    if len(word) < 4:
        words_fewer_than_4.append(word)

words_fewer_than_4
````

````python
mixed = [1, "hello", 2, "world", 3, "python"]

only_ints = []

for obj in mixed:
    if type(obj) == int:
        only_ints.append(obj)

only_ints
````
