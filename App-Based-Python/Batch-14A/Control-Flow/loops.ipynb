````python
numbers = range(1, 6)

doubled_nums = []
squared_nums = []
cube_nums = []

for num in numbers:
    doubled_nums.append(num * 2)
    squared_nums.append(num ** 2)
    cube_nums.append(num ** 3)

print(doubled_nums, squared_nums, cube_nums, sep="\n")
````

````python
import random

random_die = random.randint(1, 6)
print(random_die)

attempts = 3

while attempts > 0:
    guess = int(input("Guess the rolled die: "))
    attempts -= 1

    if guess == random_die:
        print("You won!")
        break
    else:
        print("You lost!")
        if attempts == 0:
            print("Game over, loser")
````

````python
attempts = 5
random_num = random.randint(1, 100)

while attempts > 0:
    guess = int(input("Guess the number between 1and 100"))
    attempts -= 1

    if guess > random_num:
        print(f"{guess} is too high. {attempts} trials left")
    elif guess < random_num:
        print(f"{guess} is too low. {attempts} trials left")
    else:
        print(f"Congratulations. It took you {attempts} trials. Num: {random_num}")
        break
    
    if attempts == 0:
        print(f"Game Over! The number was {random_num}")
````

* Simulates an **ATM withdrawal system** with a PIN, balance checking, and deposit/withdrawal options.


````python
print("App")

items = ['ball', 'soap', 'plate', 'perfume']

while True:
    option = input("1. View Items\n2. Add Items\n3. Quit\n")

    if option == '1':
        print(f"All items:...\n{items}\n")
    elif option == '2':
        new_item = input("Add item: ")
        items.append(new_item)
        print(f"Added <{new_item}>")
    elif option == '3':
        print("Exiting...\n")
        break
````
