```python
python_lang_dictionary = {
    'identifier': " word used to name variables, modules, functions",
    'indentation': "the space given by the left margin of a block",
    'keyword': "reserved special word built into the language"
}
query = input("Look up word in Python lnguge dictionary: ")
q = query.lower()


if q in python_lang_dictionary:
    print(f"You searched for {query}\nMeaning: {python_lang_dictionary[q]}")
else:
    print(f"'{query}' not found in dictionary")
```
