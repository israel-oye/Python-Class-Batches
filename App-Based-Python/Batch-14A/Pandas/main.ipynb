```python
df = pd.read_csv("sales.csv")

head_df = df.head()
# print(head_df)

df = df[["product", "unit_price", 'quantity', 'discount_rate']]


df['is_expensive'] = df['unit_price'] > df['unit_price'].median()


# display(df.head())
# df['unit_price'] - 100

df['gross_price'] = df['unit_price'] * df['quantity']

# display(df.head())
# print('-' * 15)

df['discounted_price'] = df['gross_price'] - (df['gross_price'] * df['discount_rate'])

df.head()

# high_value_order: one where discounted price IS ABOVE 1000
# is_cheap
#  unit_price < 20
df['high_value_order'] = df['discounted_price'] > 1000
df['is_cheap'] = df['unit_price'] < 20
df


# dicsount_category:
#   No Discount WHERE discount rate == 0
#   Low Discount WHERE discount rate < 5%
#   Medium Discount WHERE rate > 5%
#  High Discount WHERE rate > 10%

# price_tier (based on unit price)
#   Cheap < 20
#   Mid-range 20-200
#  Expensive   > 200

# is_cheap
#  unit_price < 20
```

Using the attached CSV file

Derive the following columns from the dataset:

1. Height_m — convert Height_cm to metres.


2. BMI — calculate from height and weight.


3. Weekly_Workout_Minutes — workouts per week × average workout minutes.


4. Weekly_Workout_Hours — convert weekly minutes to hours.

```python
import pandas as pd
from IPython.display import display

df = pd.read_csv('fitness_activity.csv')

# display(df.head())


df['Height_m'] = df['Height_cm'] / 100

df['BMI'] = round((df.Weight_kg / (df.Height_m ** 2)), 2)
df["Weekly_Workout_Minutes"] = df['Workouts_Per_Week'] * df['Avg_Workout_Minutes']
df['Weekly_Workout_Hours'] = round((df['Weekly_Workout_Minutes'] / 60), 2)

display(df.sample(10))

df.to_csv('fitness_activity_update.csv', index=False)
```

```python
new_df = df[['BMI', 'Weekly_Workout_Hours']]

new_df.to_csv('fitness_activity_update.csv', index=False)

new_df.to_excel('data.xlsx', index=False)
```
