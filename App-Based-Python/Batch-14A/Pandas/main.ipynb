```python
_products = [
    {"id": 1, "name": "Laptop", "price": 1500.00, "color": 4281089616},
    {"id": 2, "name": "Smartphone", "price": 950.99, "color": 4281637083},
    {"id": 3, "name": "Tablet", "price": 599.99, "color": 4280193279},
    {"id": 4, "name": "Headphones", "price": 299.99, "color": 4293348412},
    {"id": 5, "name": "Smart Watch", "price": 399.99, "color": 4294928820},
]

products_data = {
    'id': [],
    'name': [],
    'price': []
}

for p in _products:
    products_data["id"].append(p["id"])
    products_data["name"].append(p["name"])
    products_data["price"].append(p["price"])


display(products_data)
df = pd.DataFrame(products_data)
df
```

```python
df = pd.read_csv("sales.csv")

head_df = df.head()
print(head_df)

df = df[["product", "unit_price", 'quantity', 'discount_rate']]


df['is_expensive'] = df['unit_price'] > df['unit_price'].median()


# display(df.head())
# df['unit_price'] - 100

df['gross_price'] = df['unit_price'] * df['quantity']

display(df.head())
# print('-' * 15)

df['discounted_price'] = df['gross_price'] - (df['gross_price'] * df['discount_rate'])

df.head()

# high_value_order: one where discounted price IS ABOVE 1000
# is_cheap
#  unit_price < 20


# dicsount_category:
#   No Discount WHERE discount rate == 0
#   Low Discount WHERE discount rate < 5%
#   Medium Discount WHERE rate > 5%
#  High Discount WHERE rate > 10%

# price_tier (based on unit price)
#   Cheap < 20
#   Mid-range 20-200
#  Expensive   > 200

# is_cheap
#  unit_price < 20
```
