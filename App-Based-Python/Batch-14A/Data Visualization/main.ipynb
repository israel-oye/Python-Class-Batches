```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["Basit", "Ibro", "Azeezah"])

y = np.array([3, 8, 9])

# Colors: rgbckmyw
# Markers: o * . , x X + D
# Line styles: - : -- -.
plt.xlabel('Names')
plt.ylabel('Age')
plt.title("Ages of Mr. Abdul's children")

plt.grid()
plt.plot(x, y, "r:", marker="D", ms=8, mec='k')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


df = pd.read_csv('nigerian_orgs_financial_data.csv')
# display(df)

df = df.sample(3)



x_axis = df['Organization']
y_1 = df['Revenue_Billion_NGN']
y_2 = df['Operating_Expenses_Billion_NGN']


plt.plot(x_axis, y_1, 'g-', marker='*', label='Revenue')
plt.plot(x_axis, y_2, 'r-', marker='*', label='Operating expenses')
plt.xlabel('Organisation')
plt.ylabel('NGN (billions)')

plt.legend()
plt.show()

# Create a new column "Profit_Billion_NGN"
# Plot a line graph showing both Profit_Billion 
# and Tax_Paid
```

```python
df = pd.read_csv("nigerian_orgs_financial_data.csv")
# display(df)

df = df.sample(3)

df["Profit_Billion_NGN"] = df["Revenue_Billion_NGN"] - (
    df["Operating_Expenses_Billion_NGN"] + df["Tax_Paid_Billion_NGN"]
)

x_axis = df['Organization']
y_1 = df["Profit_Billion_NGN"]
y_2 = df['Tax_Paid_Billion_NGN']

plt.plot(x_axis, y_1, 'b-', marker='*', label="Profit")
plt.plot(x_axis, y_2, 'm-', marker='*', label="Tax Paid")
plt.xlabel('Organisation')
plt.ylabel('NGN (billions)')

plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('fitness_activity_update.csv').truncate(before=0, after=5)

x_axis = df['Student_ID']
y_axis = df["Avg_Workout_Minutes"]

plt.barh(x_axis, y_axis, color=['navy', 'goldenrod', 'grey'])
plt.xlabel("Student ID")
plt.ylabel("Avg Workout Duration (mins)")

# plt.grid()
plt.show()
```

```python
# Scatter plots

# Variables:

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("study_hours.csv")

x_axis = df['Hours_Studied']
y_axis = df['Test_Score']


plt.xlabel('Hours Studied')
plt.ylabel('Test Score')
plt.grid()
plt.scatter(x_axis, y_axis)

df['Hours_Studied'].corr(df['Test_Score'])
```

```python
df = pd.read_csv("study_hours.csv")

x_axis = df["Hours_Studied"]
y_axis = df["Errors_Made"]


plt.xlabel("Hours Studied")
plt.ylabel("Errors Made")
plt.grid()
plt.scatter(x_axis, y_axis, color='red')

df["Hours_Studied"].corr(df["Errors_Made"])
```
