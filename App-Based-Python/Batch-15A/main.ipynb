1. Ask the user: Enter total minutes elapsed:
2. Calculate days, hours, and minutes
3. Print the result in this format: DD:HH:MM or (DD) days, (HH) hours, (MM) minutes

```python
total_minutes_elapsed = 600_000 #int(input("Enter minutes"))

day = total_minutes_elapsed // 1440

remaining_mins_from_day = total_minutes_elapsed % 1440

hours = remaining_mins_from_day // 60

minutes = f"{remaining_mins_from_day % 60:02d}"
print(f"{total_minutes_elapsed}=> {day:02d}:{hours:02d}:{minutes}")

print(f"{4000000:3,}")
```
