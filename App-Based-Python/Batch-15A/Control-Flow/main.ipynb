```python
gender = input("Male or Female (M/F): ")
gender = gender.lower()

year_of_birth = int(input("Enter birth year: "))
age = 2026 - year_of_birth

#                    20 <= age <= 35
if gender == "m" and 1991 <= year_of_birth < 2006:
    print(f"You are eligible ✅. Age {age}\tSex: {gender}")
else:
    print(f"Sorry. Not eligible ❌ Age {age}\tSex: {gender}")
```
