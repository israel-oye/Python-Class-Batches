```python
# A = 70> B = 60> C = 50> D = 45 E = 40

score = int(input("Enter score (0-100): "))

if 70 <= score <= 100:
    print('A')
elif score >= 60:
    print('B')
elif score >= 50:
    print("C")
elif score >= 45:
    print("D")
elif score >= 40:
    print("E")
elif 0 <= score < 40:
    print("F")
else:
    print("Invalid score")
```

```python
light = 'yellow'

match light:
    case 'green':
        print("GO")
    case 'red':
        print("STOP")
    case 'yellow':
        print("GET READY TO STOP")
    case _:
        print('No match')
    
```

```python
response_code = 404

match response_code:
    case 200:
        print("OK")
    case 201:
        print("CREATED")
    case 404:
        print("NOT FOUND ERROR")
    case 500:
        print("INTERNAL SERVER ERROR")
```
