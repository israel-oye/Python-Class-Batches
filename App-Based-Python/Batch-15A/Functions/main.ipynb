Write a function called city_country() that takes in the name
of a city and its country. 
The function should return a string formatted like this:
"Santiago, Chile"

Call your function with at least three city-country pairs, and print the value that’s returned.

```python
def city_country(city: str, country: str):
    return f"{city}, {country}".title()

print(city_country('Ibadan', 'Nigeria'))
print(city_country('berlin', 'Germany'))
print(city_country('Dakar', 'SENEGAL'))

city_country_pairs = [
    ('L.A', 'USA'),
    ('Bangkok', 'Thailand'),
    ('Dublin', 'Northern Ireland'),
    ('Glasgow', 'Scotland')
]

for x, y in city_country_pairs:
    print(city_country(city=x, country=y))
```

```python
def product(*operands):
    result = 1

    for operand in operands:
        result *= operand

    return result

product(4, 7, 2)
product()
product(80, 2, 2, 2, 3)
```

Write a function that stores information about a car in a dictionary. The function should always receive a manufacturer and a model name.
It should then accept an arbitrary number of keyword arguments. 
Call the function with the required information and two other name-value pairs, such as a
color or an optional feature. 
Your function should work for a call like this one:
car = make_car('subaru', 'outback', color='blue', tow_package=True)

```python
def make_car(manufacturer: str, model: str, **kwargs):
    return {
        "manufacturer": manufacturer.title(), 
        "model": model.title(), 
        **kwargs
        }


car = make_car("subaru", "outback", color="blue", tow_package=True)
car_2 = make_car('toyota', 'slickback', weight=4, is_convertible=False, hp=3450, mileage=34_590)

car_2
```

```python
# Recursion: A function that invokes/calls itself 

# def foo():
#     foo()

def factorial(n):
    if n == 1: # base case
        return 1
    return n * factorial(n - 1)
```

```python
# Higher Order Functions
# g(f)


def executor(f):
    print("Executing function...")
    f()


# executor(f=foo)
# executor(f=bar)
# executor(f=lambda: print('Anon function...'))


def double(x):
    print(x * 2)

def executor(f):
    def wrapper(x):
        f(x)
    return wrapper

wrapper = executor(double)
wrapper(2)
```

```python
def mapp(f, collection):
    transformed_collection = []

    for item in collection:
        transformed_item = f(item)
        transformed_collection.append(transformed_item)

    return transformed_collection


mapp(lambda m: m * 100, lengths_m)


def cube(i: int):
    return i**3

mapp(cube, range(10, 50, 10))

list(map(lambda w: w.upper(), ('the', 'boy', 'is')))
```

```python
# Reduce
from functools import reduce


def total(x, y):
    return x + y


reduce(total, [5, 2, 3, 1])

reduce(total, [3, 2, 1], 10)

data = [
    {"a": 1},
    {"a": 2},
]

def total_(x, y: dict):
    return x + y['a']

reduce(total_, data, 0)
```
