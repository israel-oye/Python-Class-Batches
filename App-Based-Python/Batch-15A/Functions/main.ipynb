Write a function called city_country() that takes in the name
of a city and its country. 
The function should return a string formatted like this:
"Santiago, Chile"

Call your function with at least three city-country pairs, and print the value that’s returned.

```python
def city_country(city: str, country: str):
    return f"{city}, {country}".title()

print(city_country('Ibadan', 'Nigeria'))
print(city_country('berlin', 'Germany'))
print(city_country('Dakar', 'SENEGAL'))

city_country_pairs = [
    ('L.A', 'USA'),
    ('Bangkok', 'Thailand'),
    ('Dublin', 'Northern Ireland'),
    ('Glasgow', 'Scotland')
]

for x, y in city_country_pairs:
    print(city_country(city=x, country=y))
```
